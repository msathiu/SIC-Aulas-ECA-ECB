### Registro de decisiones de diseño
Escriban aquí, en pocas líneas, qué decidieron para cada punto de libertad y **por qué** (esto alimenta el informe).

- **Paquetes urgentes:** van en una cola aparte que siempre se atiende primero. dentro de cada cola se respeta el orden de llegada, asi ningun paquete se salta.
- **Nombres de cliente repetidos:** la busqueda por nombre devuelve todos los paquetes con ese nombre, no solo el primero, y cada uno se muestra con la cedula para distinguir a los clientes. no importa si el nombre se escribe con mayusculas.

- **Qué se puede deshacer y qué no:** se pueden deshacer las ultimas 20 acciones: recibir, atender y marcar entregado. las consultas y busquedas no se deshacen porque no cambian nada, y la carga inicial de datos del menu tampoco. si no hay historial se avisa que no hay acciones para deshacer.

### Mínimo 2 · Recepción y atención en orden de llegada
Necesitan: recibir un paquete, atender el siguiente y **ver quién sigue** sin sacarlo.

```python
# cola fifo: el primero en llegar es el primero en ser atendido
class Cola:

    def __init__(self):
        self._datos = deque()

    def encolar(self, x):
        self._datos.append(x)

    def desencolar(self):
        if self.esta_vacia():
            return None
        return self._datos.popleft()

    def frente(self):
        # ver quien sigue sin sacarlo
        if self.esta_vacia():
            return None
        return self._datos[0]

    def esta_vacia(self):
        return len(self._datos) == 0

    def __len__(self):
        return len(self._datos)

    # solo para deshacer
    def insertar_al_frente(self, x):
        self._datos.appendleft(x)

    def quitar_ultimo(self):
        if self.esta_vacia():
            return None
        return self._datos.pop()
```

```python
q = Cola()
print(q.desencolar())        # cola vacia
for x in ["a", "b", "c"]:
    q.encolar(x)
print(q.frente(), len(q))    # ver quien sigue sin sacarlo
print(q.desencolar(), q.frente())
q.insertar_al_frente("a")
print(q.frente())
print(q.quitar_ultimo(), len(q))
```

### Mínimo 3 · Consulta por código de seguimiento (tabla hash propia)
Implementen la tabla hash completa: función hash, insertar, buscar y manejo de colisiones. **Sin `dict` ni `set`.**

```python
# tabla hash propia con encadenamiento: cada cubeta es una lista de pares clave y valor

def es_primo(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True


def siguiente_primo(n):
    # primer numero primo desde n
    while not es_primo(n):
        n += 1
    return n


class TablaHash:
    def __init__(self, capacidad=CAPACIDAD_HASH):
        self.capacidad = capacidad
        self.cantidad = 0
        self.tabla = [[] for _ in range(capacidad)]

    def _hash(self, clave):
        # por cada letra: valor por 31 mas su codigo ascii, modulo la capacidad
        valor = 0
        for c in clave:
            valor = (valor * BASE_HASH + ord(c)) % self.capacidad
        return valor

    def _buscar_par(self, clave):
        # solo revisa la cubeta donde cae la clave
        comparaciones = 0
        for par in self.tabla[self._hash(clave)]:
            comparaciones += 1
            if par[0] == clave:
                return par, comparaciones
        return None, comparaciones

    def insertar(self, clave, valor):
        if self._buscar_par(clave)[0] is not None:
            return False                       # codigo repetido
        # si hay colision el par se agrega a la misma lista
        self.tabla[self._hash(clave)].append([clave, valor])
        self.cantidad += 1
        if self.factor_carga() > CARGA_MAXIMA:
            self._rehash()
        return True

    def buscar(self, clave):
        par, comparaciones = self._buscar_par(clave)
        if par is None:
            return None, comparaciones
        return par[1], comparaciones

    def eliminar(self, clave):
        cubeta = self.tabla[self._hash(clave)]
        for i in range(len(cubeta)):
            if cubeta[i][0] == clave:
                cubeta.pop(i)
                self.cantidad -= 1
                return True
        return False

    def factor_carga(self):
        return self.cantidad / self.capacidad

    def _rehash(self):
        # crea una tabla mas grande y vuelve a colocar todos los pares
        anterior = self.tabla
        self.capacidad = siguiente_primo(2 * self.capacidad)
        self.tabla = [[] for _ in range(self.capacidad)]
        self.cantidad = 0
        for cubeta in anterior:
            for clave, valor in cubeta:
                self.insertar(clave, valor)

    def __len__(self):
        return self.cantidad
```

```python
t = TablaHash(7)                                 # tabla chica para provocar colisiones y rehash
print(t.buscar("ENV-00000"))                     # tabla vacia
for i in range(1, 7):
    t.insertar(f"ENV-{i:05d}", f"paquete {i}")
print(t.insertar("ENV-00001", "otro"))           # codigo duplicado
print(t.buscar("ENV-00003")[0])
print(t.capacidad, round(t.factor_carga(), 2))   # hubo rehash a 17
print(t.eliminar("ENV-00003"), t.buscar("ENV-00003")[0])
print([len(c) for c in t.tabla])                 # paquetes en cada cubeta
```

### Mínimo 4 · Deshacer acciones
Decidan qué acciones se registran y **qué información hay que guardar** para poder revertirlas.

```python
# cada accion guarda lo necesario para revertirla:
# el tipo, el paquete y la cola o el estado anterior
class Accion:
    def __init__(self, tipo, paquete, anterior=None):
        self.tipo = tipo
        self.paquete = paquete
        self.anterior = anterior

    def __str__(self):
        return f"{self.tipo} {self.paquete.codigo}"
```

```python
# pila lifo: la ultima accion es la primera que se deshace
class Pila:
    def __init__(self, limite=LIMITE_DESHACER):
        self._datos = []
        self.limite = limite

    def apilar(self, accion):
        self._datos.append(accion)
        if len(self._datos) > self.limite:
            self._datos.pop(0)          # si pasa el limite se olvida la mas antigua

    def desapilar(self):
        if self.esta_vacia():
            return None
        return self._datos.pop()

    def cima(self):
        if self.esta_vacia():
            return None
        return self._datos[-1]

    def esta_vacia(self):
        return len(self._datos) == 0

    def __len__(self):
        return len(self._datos)
```

```python
pila = Pila(limite=3)
print(pila.desapilar())                            # deshacer sin historial
q = Cola()
q.encolar(Paquete("ENV-1", Cliente("ana torres", "1"), 101, 1.0))
atendido = q.desencolar()
atendido.estado = EN_RUTA
pila.apilar(Accion("atender", atendido, q))
a = pila.desapilar()                               # deshacer atender: vuelve al frente de su cola
a.anterior.insertar_al_frente(a.paquete)
a.paquete.estado = EN_ESPERA
print(q.frente().codigo, q.frente().estado)
for i in range(5):
    pila.apilar(i)
print(len(pila), pila.cima())                      # solo guarda las ultimas 3
```

### Mínimo 5 · Búsqueda de zona por código (búsqueda binaria)
La lista de zonas debe estar **ordenada** antes de buscar.

```python
# lista de zonas ordenada por codigo y busqueda binaria
def busqueda_binaria(lista, clave, key=lambda x: x):
    bajo, alto = 0, len(lista) - 1
    comparaciones = 0
    while bajo <= alto:
        mitad = (bajo + alto) // 2
        comparaciones += 1
        valor = key(lista[mitad])
        if valor == clave:
            return mitad, comparaciones
        if valor < clave:
            bajo = mitad + 1      # descarta la mitad izquierda
        else:
            alto = mitad - 1      # descarta la mitad derecha
    return -1, comparaciones


def posicion_de_insercion(lista, clave, key=lambda x: x):
    # misma idea de la binaria: devuelve donde debe ir la clave para no romper el orden
    bajo, alto = 0, len(lista) - 1
    while bajo <= alto:
        mitad = (bajo + alto) // 2
        if key(lista[mitad]) < clave:
            bajo = mitad + 1
        else:
            alto = mitad - 1
    return bajo


class ListaZonas:
    def __init__(self, zonas=()):
        self._zonas = []
        for zona in zonas:       # cada zona entra directo en su posicion
            self.agregar(zona)

    def buscar(self, codigo):
        pos, comparaciones = busqueda_binaria(self._zonas, codigo, key=lambda z: z.codigo)
        if pos == -1:
            return None, comparaciones
        return self._zonas[pos], comparaciones

    def agregar(self, zona):
        if self.buscar(zona.codigo)[0] is not None:
            return False         # codigo de zona repetido
        pos = posicion_de_insercion(self._zonas, zona.codigo, key=lambda z: z.codigo)
        self._zonas.insert(pos, zona)
        return True

    def codigos(self):
        return [z.codigo for z in self._zonas]

    def __len__(self):
        return len(self._zonas)
```

```python
zonas = ListaZonas([Zona(330, "centro", "santo domingo"), Zona(101, "centro", "quito"),
                    Zona(210, "sur", "guayaquil"), Zona(105, "norte", "quito")])
print(zonas.codigos())                      # quedaron ordenadas al cargar
z, comps = zonas.buscar(210)
print(z, " comparaciones:", comps)
print(zonas.buscar(999))                    # zona inexistente
print(zonas.agregar(Zona(150, "valle", "quito")), zonas.agregar(Zona(150, "otra", "quito")))
print(zonas.codigos())                      # la zona nueva quedo en su lugar
print(busqueda_binaria([3, 8, 15, 21, 40], 21))
```

### Mínimo 6 · Búsqueda de paquetes por nombre de cliente (búsqueda secuencial)
Piensen qué devuelve su búsqueda si hay varios paquetes con el mismo nombre.

```python
# busqueda secuencial por nombre: revisa todos los paquetes y devuelve todos los que coinciden
def buscar_paquetes_por_cliente(paquetes, nombre):
    resultados = []
    comparaciones = 0

    for paquete in paquetes:
        comparaciones += 1
        if paquete.cliente.nombre.lower() == nombre.lower():
            resultados.append(paquete)

    return resultados, comparaciones
```

```python
lista = [
    Paquete("ENV-001", Cliente("ana torres", "1"), 101, 1.5),
    Paquete("ENV-002", Cliente("pedro ruiz", "2"), 210, 2.0),
    Paquete("ENV-003", Cliente("ana torres", "3"), 105, 0.8),
]

encontrados, comps = buscar_paquetes_por_cliente(lista, "ana torres")   # nombre repetido

for p in encontrados:
    print(p)

print("comparaciones:", comps)
```

### sistema integrado
la clase centro une todo lo anterior: una cola para urgentes y otra para normales, la tabla hash para consultar por codigo, la pila para deshacer, la lista ordenada de zonas y la busqueda por nombre. los paquetes que llegan sin codigo reciben un codigo interno que empieza con SC para poder guardarlos en la tabla hash. como el cliente no conoce ese codigo, esos paquetes se buscan por nombre.

```python
# centro de distribucion: une la cola, la tabla hash, la pila, las zonas y la busqueda por nombre
class Centro:
    def __init__(self, zonas):
        self.cola_urgente = Cola()
        self.cola_normal = Cola()
        self.tabla = TablaHash()          # consulta por codigo
        self.historial = Pila()           # acciones que se pueden deshacer
        self.zonas = ListaZonas(zonas)    # zonas ordenadas por codigo
        self.registro = []                # todos los paquetes, para buscar por nombre
        self.sin_codigo = 0               # cuantos paquetes llegaron sin codigo

    def recibir(self, paquete):
        if self.zonas.buscar(paquete.zona)[0] is None:
            return f"la zona {paquete.zona} no existe"
        if paquete.codigo == "":
            self.sin_codigo += 1
            paquete.codigo = f"SC-{self.sin_codigo:04d}"
        if not self.tabla.insertar(paquete.codigo, paquete):
            return f"el codigo {paquete.codigo} ya existe"
        paquete.orden = len(self.registro) + 1
        cola = self.cola_urgente if paquete.urgente else self.cola_normal
        cola.encolar(paquete)
        self.registro.append(paquete)
        self.historial.apilar(Accion("recibir", paquete, cola))
        return f"paquete {paquete.codigo} recibido"

    def cola_actual(self):
        # los urgentes van primero, si no hay urgentes se atiende la cola normal
        if self.cola_urgente.esta_vacia():
            return self.cola_normal
        return self.cola_urgente

    def siguiente(self):
        return self.cola_actual().frente()

    def atender(self):
        cola = self.cola_actual()
        paquete = cola.desencolar()
        if paquete is not None:
            paquete.estado = EN_RUTA
            self.historial.apilar(Accion("atender", paquete, cola))
        return paquete

    def entregar(self, codigo):
        paquete, _ = self.tabla.buscar(codigo)
        if paquete is None:
            return f"no existe un paquete con el codigo {codigo}"
        if paquete.estado != EN_RUTA:
            return f"el paquete {codigo} no esta en ruta"
        paquete.estado = ENTREGADO
        self.historial.apilar(Accion("entregar", paquete, EN_RUTA))
        return f"paquete {codigo} entregado"

    def deshacer(self):
        accion = self.historial.desapilar()
        if accion is None:
            return "no hay acciones para deshacer"
        paquete = accion.paquete
        if accion.tipo == "recibir":       # sale de su cola, de la tabla y del registro
            accion.anterior.quitar_ultimo()
            self.tabla.eliminar(paquete.codigo)
            self.registro.pop()
        elif accion.tipo == "atender":     # vuelve al frente de su cola
            accion.anterior.insertar_al_frente(paquete)
            paquete.estado = EN_ESPERA
        else:                              # entregar: vuelve al estado anterior
            paquete.estado = accion.anterior
        return f"accion deshecha: {accion}"

    def consultar(self, codigo):
        return self.tabla.buscar(codigo)

    def buscar_por_nombre(self, nombre):
        return buscar_paquetes_por_cliente(self.registro, nombre)

    def buscar_zona(self, codigo):
        return self.zonas.buscar(codigo)

    def agregar_zona(self, zona):
        return self.zonas.agregar(zona)
```

### Mínimo 8 · Medición de comparaciones
Midan cuántas comparaciones hace cada búsqueda (secuencial, binaria y hash) con **100, 1000 y 10000** datos y muestren una tabla con los resultados. Después escriban en la celda siguiente qué observan.

```python
# medicion: cada prueba busca el ultimo dato, que es el peor caso para la busqueda secuencial
def busqueda_secuencial(lista, objetivo):
    comparaciones = 0
    for elemento in lista:
        comparaciones += 1
        if elemento == objetivo:
            return comparaciones
    return comparaciones


def prueba_secuencial(n):
    datos = list(range(n))
    return busqueda_secuencial(datos, n - 1)


def prueba_binaria(n):
    datos = list(range(n))
    _, comps = busqueda_binaria(datos, n - 1)
    return comps


def prueba_hash(n):
    tabla = TablaHash()
    for i in range(n):
        tabla.insertar(f"ENV-{i:05d}", i)
    _, comps = tabla.buscar(f"ENV-{n - 1:05d}")
    return comps
```

```python
# tabla de resultados, tambien se usa en el menu
def mostrar_medicion():
    print(f"{'datos':>6} {'secuencial':>11} {'binaria':>8} {'hash':>5}")
    for n in (100, 1000, 10000):
        print(f"{n:>6} {prueba_secuencial(n):>11} {prueba_binaria(n):>8} {prueba_hash(n):>5}")


mostrar_medicion()
```

### Análisis de las mediciones
¿Qué pasa con las comparaciones al multiplicar los datos por 10? ¿Cuál búsqueda crece más rápido? ¿Su tabla hash se comportó como esperaban?

al multiplicar los datos por 10, la busqueda secuencial tambien multiplica sus comparaciones por 10: pasa de 100 a 1000 y a 10000, porque revisa elemento por elemento hasta llegar al ultimo. es la que crece mas rapido.

la busqueda binaria crece muy lento: pasa de 7 a 10 y luego a 14 comparaciones, porque en cada paso descarta la mitad de los datos. por eso la usamos para las zonas, que estan ordenadas.

la tabla hash se comporto como esperabamos: necesito una sola comparacion con 100, 1000 y 10000 datos. cuando dos codigos caen en la misma cubeta hace alguna comparacion extra, pero como la tabla crece al pasar del 75 por ciento de ocupacion, las cubetas se mantienen cortas.

por eso usamos la tabla hash para consultar por codigo de seguimiento, la binaria para las zonas y la secuencial para buscar por nombre, porque recorre todos los paquetes y devuelve todos los que tienen ese nombre.

### extra: menu por consola
ejecuten esta celda para usar todo el sistema desde un menu. empieza con las 5 zonas y los 20 paquetes de ejemplo, y esa carga inicial no se puede deshacer.

```python
# menu por consola: usa el centro de distribucion que une todas las estructuras

def leer_texto(mensaje):
    texto = input(mensaje).strip()
    while texto == "":
        texto = input("este dato es obligatorio: ").strip()
    return texto


def leer_entero(mensaje):
    dato = input(mensaje).strip()
    while not dato.isdigit():
        dato = input("escriba solo numeros: ").strip()
    return int(dato)


def leer_peso(mensaje):
    while True:
        try:
            peso = float(input(mensaje).replace(",", "."))    # acepta 2,5 o 2.5
            if peso > 0:
                return peso
        except ValueError:
            pass
        print("escriba un peso mayor que cero")


def mostrar(dato, mensaje):
    # muestra el resultado o un mensaje si no se encontro nada
    if dato is None:
        print(mensaje)
    else:
        print(dato)


def pausa():
    input("\npresione enter para continuar")
    clear_output()


def mostrar_opciones():
    print("centro de distribucion envirapido\n")
    print("1. recibir paquete")
    print("2. ver quien sigue")
    print("3. atender siguiente paquete")
    print("4. marcar paquete como entregado")
    print("5. consultar por codigo")
    print("6. buscar por nombre del cliente")
    print("7. buscar zona")
    print("8. agregar zona")
    print("9. deshacer ultima accion")
    print("10. ver todos los paquetes")
    print("11. comparar busquedas")
    print("0. salir\n")


def menu(centro):
    while True:
        mostrar_opciones()
        opcion = input("elija una opcion: ").strip()
        clear_output()

        if opcion == "1":
            nombre = leer_texto("nombre del cliente: ").lower()
            cedula = input("cedula, enter si no tiene: ").strip()
            if cedula == "":
                cedula = "sin registrar"
            codigo = input("codigo de seguimiento, enter si no tiene: ").strip().upper()
            zona = leer_entero("codigo de zona: ")
            peso = leer_peso("peso en kg: ")
            urgente = input("es urgente, escriba s para si: ").strip().lower() == "s"
            print(centro.recibir(Paquete(codigo, Cliente(nombre, cedula), zona, peso, urgente)))

        elif opcion == "2":
            print("paquete que sigue:")
            mostrar(centro.siguiente(), "no hay paquetes en espera")

        elif opcion == "3":
            print("paquete atendido:")
            mostrar(centro.atender(), "no hay paquetes en espera")

        elif opcion == "4":
            print(centro.entregar(leer_texto("codigo del paquete: ").upper()))

        elif opcion == "5":
            paquete, comps = centro.consultar(leer_texto("codigo del paquete: ").upper())
            mostrar(paquete, "no existe un paquete con ese codigo")
            print("comparaciones:", comps)

        elif opcion == "6":
            encontrados, comps = centro.buscar_por_nombre(leer_texto("nombre del cliente: "))
            if len(encontrados) == 0:
                print("no hay paquetes con ese nombre")
            for p in encontrados:
                print(f"{p}  cedula {p.cliente.cedula}")
            print("comparaciones:", comps)

        elif opcion == "7":
            zona, comps = centro.buscar_zona(leer_entero("codigo de zona: "))
            mostrar(zona, "esa zona no existe")
            print("comparaciones:", comps)

        elif opcion == "8":
            codigo = leer_entero("codigo de la zona nueva: ")
            nombre = leer_texto("nombre de la zona: ").lower()
            ciudad = leer_texto("ciudad: ").lower()
            if centro.agregar_zona(Zona(codigo, nombre, ciudad)):
                print("zona agregada, zonas actuales:", centro.zonas.codigos())
            else:
                print("esa zona ya existe")

        elif opcion == "9":
            print(centro.deshacer())

        elif opcion == "10":
            print("urgentes en espera:", len(centro.cola_urgente))
            print("normales en espera:", len(centro.cola_normal))
            for p in centro.registro:
                print(p)

        elif opcion == "11":
            mostrar_medicion()

        elif opcion == "0":
            print("sistema cerrado")
            break

        else:
            print("opcion no valida")

        pausa()


# datos de ejemplo: las zonas y los 20 paquetes del minimo 7
centro_menu = Centro(zonas_ejemplo)
for i in range(20):
    cliente = Cliente(f"cliente{i + 1}", str(1000 + i))
    centro_menu.recibir(Paquete(f"ENV-{i + 1:05d}", cliente, 100 + i % 5, round((i + 1) * 0.5, 1)))
centro_menu.historial = Pila()    # la carga inicial no se puede deshacer

menu(centro_menu)
```
