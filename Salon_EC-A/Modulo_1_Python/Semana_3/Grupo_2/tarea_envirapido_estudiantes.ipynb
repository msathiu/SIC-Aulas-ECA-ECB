### Mínimo 2 · Recepción y atención en orden de llegada
Necesitan: recibir un paquete, atender el siguiente y **ver quién sigue** sin sacarlo.

```python
# ===== MÍNIMO 2 · COLA DE ORDEN DE LLEGADA (FIFO) =====
# Estructura elegida: COLA implementada con lista enlazada (un puntero al frente y otro al final).
# Así recibir y atender cuestan O(1): no hay que recorrer ni mover elementos.

class NodoCola:
    def __init__(self, paquete):
        self.paquete = paquete
        self.siguiente = None


class ColaLlegada:
    def __init__(self):
        self.frente = None    # el que se atiende primero
        self.final = None     # el último que llegó
        self.tamano = 0

    def esta_vacia(self):
        return self.frente is None

    def recibir(self, paquete):
        """Un paquete llega: se coloca al final de la cola."""
        nuevo = NodoCola(paquete)
        if self.esta_vacia():
            self.frente = nuevo
        else:
            self.final.siguiente = nuevo
        self.final = nuevo
        self.tamano += 1

    def atender(self):
        """Saca y devuelve el paquete que lleva más tiempo esperando. Si no hay, devuelve None."""
        if self.esta_vacia():
            return None
        paquete = self.frente.paquete
        self.frente = self.frente.siguiente
        if self.frente is None:       # la cola quedó vacía: también se limpia el final
            self.final = None
        self.tamano -= 1
        return paquete

    def ver_siguiente(self):
        """Muestra quién sigue SIN sacarlo de la cola. Si no hay, devuelve None."""
        if self.esta_vacia():
            return None
        return self.frente.paquete


# --- Prueba rápida ---
cola = ColaLlegada()
print("Cola vacía, ver_siguiente:", cola.ver_siguiente())
print("Cola vacía, atender:", cola.atender())
for i, nombre in enumerate(["Ana", "Luis", "Marta"], start=1):
    cola.recibir(Paquete(f"EV-{i:04d}", Cliente(nombre), 101))
print("Sigue:", cola.ver_siguiente(), "| en cola:", cola.tamano)
print("Atendido:", cola.atender())
print("Atendido:", cola.atender())
print("Sigue:", cola.ver_siguiente(), "| en cola:", cola.tamano)
```

### Mínimo 3 · Consulta por código de seguimiento (tabla hash propia)
Implementen la tabla hash completa: función hash, insertar, buscar y manejo de colisiones. **Sin `dict` ni `set`.**

```python
# ===== MÍNIMO 3 · TABLA HASH PROPIA (sin dict ni set) =====
# Decisiones:
#  - Tamaño inicial: 101 (número primo, reparte mejor los índices).
#  - Función hash: polinómica sobre los caracteres del código -> h = (h*31 + ord(c)) % tamaño
#  - Colisiones: ENCADENAMIENTO (cada casilla guarda una lista enlazada de nodos).
#  - Tabla "llena": si factor de carga (elementos/tamaño) > 0.75 se hace REHASH:
#    se crea una tabla de tamaño doble+1 y se reinsertan todos los elementos.

class NodoHash:
    def __init__(self, codigo, paquete):
        self.codigo = codigo
        self.paquete = paquete
        self.siguiente = None


class TablaHash:
    def __init__(self, tamano=101):
        self.tamano = tamano
        self.casillas = [None] * tamano    # cada casilla es el inicio de una cadena
        self.cantidad = 0
        self.comparaciones = 0             # contador para la medición del Mínimo 8

    def _hash(self, codigo):
        h = 0
        for caracter in codigo:
            h = (h * 31 + ord(caracter)) % self.tamano
        return h

    def insertar(self, codigo, paquete):
        """Inserta el paquete. Si el código ya existe, lo reemplaza y devuelve False.
        Devuelve True si fue un código nuevo."""
        if self.cantidad / self.tamano > 0.75:
            self._rehash()
        indice = self._hash(codigo)
        actual = self.casillas[indice]
        while actual is not None:               # revisa la cadena (colisión o repetido)
            if actual.codigo == codigo:
                actual.paquete = paquete
                return False
            actual = actual.siguiente
        nuevo = NodoHash(codigo, paquete)       # se agrega al inicio de la cadena
        nuevo.siguiente = self.casillas[indice]
        self.casillas[indice] = nuevo
        self.cantidad += 1
        return True

    def buscar(self, codigo):
        """Devuelve el paquete con ese código o None si no existe.
        Cuenta una comparación por cada nodo revisado."""
        actual = self.casillas[self._hash(codigo)]
        while actual is not None:
            self.comparaciones += 1
            if actual.codigo == codigo:
                return actual.paquete
            actual = actual.siguiente
        return None

    def eliminar(self, codigo):
        """Quita el código de la tabla. Devuelve True si existía (útil para deshacer)."""
        indice = self._hash(codigo)
        actual = self.casillas[indice]
        anterior = None
        while actual is not None:
            if actual.codigo == codigo:
                if anterior is None:
                    self.casillas[indice] = actual.siguiente
                else:
                    anterior.siguiente = actual.siguiente
                self.cantidad -= 1
                return True
            anterior = actual
            actual = actual.siguiente
        return False

    def _rehash(self):
        """Agranda la tabla y reubica todos los elementos (los índices cambian con el nuevo tamaño)."""
        viejas = self.casillas
        self.tamano = self.tamano * 2 + 1
        self.casillas = [None] * self.tamano
        self.cantidad = 0
        for inicio in viejas:
            actual = inicio
            while actual is not None:
                self.insertar(actual.codigo, actual.paquete)
                actual = actual.siguiente

    def factor_de_carga(self):
        return self.cantidad / self.tamano


# --- Prueba rápida ---
tabla = TablaHash(tamano=5)    # tabla pequeña a propósito para forzar colisiones y rehash
for i in range(1, 9):
    cod = f"EV-{i:04d}"
    tabla.insertar(cod, Paquete(cod, Cliente(f"Cliente {i}"), 101))
print("Tamaño tras rehash:", tabla.tamano, "| elementos:", tabla.cantidad)
print("Buscar EV-0003:", tabla.buscar("EV-0003"))
print("Buscar inexistente:", tabla.buscar("XX-9999"))
print("Repetido devuelve:", tabla.insertar("EV-0003", Paquete("EV-0003", Cliente("Otro"), 5)))
print("Eliminar EV-0003:", tabla.eliminar("EV-0003"), "| buscar de nuevo:", tabla.buscar("EV-0003"))
# Colisión demostrada: dos códigos distintos en la misma casilla
t2 = TablaHash(tamano=3)
t2.insertar("A", Paquete("A", Cliente("x"), 1)); t2.insertar("D", Paquete("D", Cliente("y"), 1))
print("Mismo índice 'A' y 'D':", t2._hash("A") == t2._hash("D"), "| ambos se encuentran:", t2.buscar("A") is not None, t2.buscar("D") is not None)
```

### Mínimo 4 · Deshacer acciones
Decidan qué acciones se registran y **qué información hay que guardar** para poder revertirlas.

```python
# Tu código aquí: historial y deshacer


class NodoPila:
    def __init__(self, tipo, datos):
        self.tipo = tipo          # ej: "recibir_paquete", "cambiar_estado"
        self.datos = datos        # diccionario o tupla con la info para revertir
        self.siguiente = None


class PilaDeshacer:
    def __init__(self):
        self.tope = None
        self.tamano = 0

    def esta_vacia(self):
        return self.tope is None

    def apilar(self, tipo, datos):
        """Registra una nueva acción en la pila."""
        nuevo = NodoPila(tipo, datos)
        nuevo.siguiente = self.tope
        self.tope = nuevo
        self.tamano += 1

    def desapilar(self):
        """Saca y devuelve la última acción registrada."""
        if self.esta_vacia():
            return None
        accion = self.tope
        self.tope = self.tope.siguiente
        self.tamano -= 1
        return accion


# --- Prueba rápida del Mínimo 4 ---
pila_acciones = PilaDeshacer()

# Simulamos registrar acciones
pila_acciones.apilar("recibir", {"codigo": "EV-0001", "paquete": p})
pila_acciones.apilar("entregar", {"codigo": "EV-0001", "estado_anterior": "en_espera"})

print("Acciones en pila:", pila_acciones.tamano)
ultima = pila_acciones.desapilar()
print(f"Deshaciendo última acción -> Tipo: {ultima.tipo}, Datos: {ultima.datos}")
print("Acciones restantes en pila:", pila_acciones.tamano)
```

### Mínimo 5 · Búsqueda de zona por código (búsqueda binaria)
La lista de zonas debe estar **ordenada** antes de buscar.

```python
# Tu código aquí: lista de zonas ordenada y búsqueda binaria

class GestorZonas:
    def __init__(self):
        self.zonas = []   # Usamos una lista interna pero ordenamos manualmente

    def agregar_zona(self, zona):
        """Agrega una zona y asegura que el arreglo se mantenga ordenado."""
        self.zonas.append(zona)
        self._ordenar_por_codigo()

    def _ordenar_por_codigo(self):
        """Ordenamiento por inserción O(n^2) ideal para listas pequeñas/medianas."""
        for i in range(1, len(self.zonas)):
            actual = self.zonas[i]
            j = i - 1
            while j >= 0 and self.zonas[j].codigo > actual.codigo:
                self.zonas[j + 1] = self.zonas[j]
                j -= 1
            self.zonas[j + 1] = actual

    def buscar_zona_binaria(self, codigo_buscado):
        """
        Búsqueda binaria iterativa. 
        Retorna el objeto Zona si lo encuentra, o None si no existe.
        Costo temporal: O(log n).
        """
        inicio = 0
        fin = len(self.zonas) - 1

        while inicio <= fin:
            medio = (inicio + fin) // 2
            if self.zonas[medio].codigo == codigo_buscado:
                return self.zonas[medio]
            elif self.zonas[medio].codigo < codigo_buscado:
                inicio = medio + 1
            else:
                fin = medio - 1

        return None


# --- Prueba rápida del Mínimo 5 ---
gestor = GestorZonas()
# Agregamos en desorden a propósito para probar el ordenamiento automático
gestor.agregar_zona(Zona(300, "Sur"))
gestor.agregar_zona(Zona(100, "Norte"))
gestor.agregar_zona(Zona(200, "Centro"))

print("Zonas ordenadas automáticamente:")
for z in gestor.zonas:
    print(z)

# Búsqueda existente
encontrada = gestor.buscar_zona_binaria(200)
print("\nBuscar zona código 200:", encontrada)

# Búsqueda inexistente
no_encontrada = gestor.buscar_zona_binaria(999)
print("Buscar zona código 999:", no_encontrada)
```

### Mínimo 6 · Búsqueda de paquetes por nombre de cliente (búsqueda secuencial)
Piensen qué devuelve su búsqueda si hay varios paquetes con el mismo nombre.

```python
# Tu código aquí: búsqueda secuencial por nombre
def buscar_por_nombre(paquetes, nombre):
    """Búsqueda secuencial: devuelve (lista de coincidencias, comparaciones)."""
    resultados = []
    comparaciones = 0

    for paquete in paquetes:
        comparaciones += 1                         # cada paquete revisado cuenta
        if nombre == paquete.cliente.nombre:       # ¿coincide el nombre del cliente?
            resultados.append(paquete)             # si coincide, lo guardo

    return resultados, comparaciones

def generar_paquetes(n, max_nombre=10000):
    paquetes = []
    for i in range(n):
        codigo = f"EV-{i:05d}"                              # texto único: EV-00000, EV-00001...
        nombre = "Cliente" + str(randint(1, max_nombre))
        paquetes.append(Paquete(codigo, Cliente(nombre), 101))
    return paquetes

lista = generar_paquetes(20, 5)    # pocos nombres posibles: habrá repetidos
res, comps = buscar_por_nombre(lista, "Cliente3")
print(len(res), comps)
```

### Mínimo 8 · Medición de comparaciones
Midan cuántas comparaciones hace cada búsqueda (secuencial, binaria y hash) con **100, 1000 y 10000** datos y muestren una tabla con los resultados. Después escriban en la celda siguiente qué observan.

```python
# Tu código aquí: mediciones
# ===== MÍNIMO 8 · MEDICIÓN DE COMPARACIONES =====
def busqueda_binaria_contando(zonas, codigo):
    """Igual que la binaria del grupo, pero cuenta las comparaciones."""
    inicio, fin = 0, len(zonas) - 1
    comparaciones = 0
    while inicio <= fin:
        medio = (inicio + fin) // 2
        comparaciones += 1                         # una comparación por vuelta
        if zonas[medio].codigo == codigo:
            return zonas[medio], comparaciones
        elif zonas[medio].codigo < codigo:
            inicio = medio + 1
        else:
            fin = medio - 1
    return None, comparaciones

print(f"{'Datos':>7} | {'Secuencial':>10} | {'Binaria':>8} | {'Hash (promedio)':>15}")
print("-" * 50)

for n in [100, 1000, 10000]:
    paquetes = generar_paquetes(n)

    # Secuencial: peor caso (nombre que no existe -> recorre todo)
    _, c_sec = buscar_por_nombre(paquetes, "NoExiste")

    # Binaria: peor caso (código que no existe). Lista de zonas ya ordenada.
    zonas = [Zona(i, f"Zona{i}") for i in range(n)]
    _, c_bin = busqueda_binaria_contando(zonas, -1)

    # Hash: promedio de buscar TODOS los códigos que sí existen
    tabla = TablaHash()
    for p in paquetes:
        tabla.insertar(p.codigo, p)
    tabla.comparaciones = 0                        # insertar no cuenta, pero así empezamos limpios
    for p in paquetes:
        tabla.buscar(p.codigo)
    c_hash = tabla.comparaciones / n

    print(f"{n:>7} | {c_sec:>10} | {c_bin:>8} | {c_hash:>15.2f}")
```
