- **Paquetes urgentes:** Decidimos utilizar una cola de prioridad con tres niveles: 1 alta, 2 media y 3 baja. Así, los paquetes más urgentes se atienden primero, pero se mantiene el orden de llegada dentro de cada nivel. También se permite cambiar la prioridad mientras el paquete espera, conservando su orden original.
- **Nombres de cliente repetidos:** Decidimos utilizar únicamente el código de seguimiento como clave de la tabla hash, ya que los nombres pueden repetirse. La búsqueda por nombre devuelve todos los paquetes que coinciden, sin importar mayúsculas o espacios adicionales. Además, se muestra el teléfono para diferenciar a los clientes.

-  **Paquetes sin código:** Decidimos asignar None al código de seguimiento cuando no está disponible. Estos paquetes se mantienen en la lista general y en la cola, pero no se guardan en la tabla hash. Para encontrarlos, se utiliza el nombre del cliente.

- **Tamaño y función hash / tabla llena:** Decidimos iniciar con una tabla de 11 casillas y utilizar una función hash polinómica, donde cada carácter influye en la posición del código. Esto permite distribuir mejor los elementos y reducir las colisiones. Las colisiones se resuelven mediante listas en cada casilla. Cuando la ocupación promedio supera el 75 %, duplicamos el tamaño de la tabla y volvemos a insertar los elementos para calcular sus nuevas posiciones.
- **Qué se puede deshacer y qué no:** Decidimos permitir deshacer las acciones de atender, entregar y cambiar la prioridad, pero no la recepción de paquetes. Esto permite corregir errores, como entregar un paquete por equivocación. Para hacerlo, utilizamos una pila, de manera que la última acción realizada sea la primera en deshacerse. Los paquetes pasan por los estados de espera, atendido y entregado.
- **Zonas (¿cambian? ¿cómo se mantienen ordenadas?):** Decidimos mantener la lista de zonas fija durante el funcionamiento del sistema, porque están asociadas a códigos numéricos constantes. Se ordenan una sola vez al inicio y se utiliza búsqueda binaria para encontrar una zona de manera más eficiente.
- **Cómo se informa que algo no se encontró:** Decidimos que las estructuras de datos no muestren mensajes directamente, sino que devuelvan None o una lista vacía cuando no encuentran un elemento. De esta manera, el sistema puede informar al usuario con un mensaje como "El paquete ### no se encontró". En la tabla hash, también se verifica que el código buscado no esté dentro de la lista de elementos de la casilla correspondiente.

### Mínimo 1 · Modelado: paquete, cliente y zona
Definan las clases o estructuras. Antes de programar, escriban en papel los **sustantivos** y los **verbos** del problema.

**Justificación del modelado**
- `Paquete` guarda una referencia a su `Cliente` y a su `Zona`, no copias de sus datos.
- `codigo` puede ser `None`: así se modelan los paquetes que llegan sin código de seguimiento.
- `estado` (espera, atendido, entregado) permite saber en qué etapa está cada paquete y validar acciones (por ejemplo, no entregar algo que nunca se atendió).
- `prioridad` va de 1 (alta) a 3 (baja) y vale 3 por defecto, porque la mayoría de los paquetes son normales.
- `orden` guarda el número de llegada: sirve de desempate en la cola y para devolver un paquete a su lugar original al deshacer.


```python
class Cliente:
  def __init__(self, nombre, telefono):
    self.nombre = nombre
    self.telefono = telefono

  def __str__(self):
    return f'Nombre: {self.nombre} | Telefono: {self.telefono}'

class Zona:
  def __init__(self, codigo, nombre):
    self.codigo = codigo
    self.nombre = nombre

  def __str__(self):
    return f'Zona: {self.nombre} | Codigo: {self.codigo}'

class Paquete:
  # codigo: puede ser None (paquete que llegó sin código de seguimiento)
  # prioridad: 1 = alta, 2 = media, 3 = baja (valor por defecto)
  # estado: 'espera', 'atendido', 'entregado'
  def __init__(self, codigo, cliente, zona, descripcion, prioridad=3):
    if prioridad not in (1, 2, 3):
      raise ValueError('La prioridad debe ser 1 (alta), 2 (media) o 3 (baja)')
    self.codigo = codigo
    self.cliente = cliente
    self.zona = zona
    self.descripcion = descripcion
    self.prioridad = prioridad
    self.estado = 'espera'
    self.orden = None   # número de llegada: lo asigna la cola al recibir el paquete

  def __str__(self):
    codigo = self.codigo if self.codigo is not None else 'SIN CODIGO'
    return f'Paquete: {codigo} con prioridad {self.prioridad} para {self.cliente.nombre} | Zona: {self.zona.nombre} - {self.zona.codigo} | Estado: {self.estado}'
```

### Mínimo 2 · Recepción y atención en orden de llegada
Necesitan: recibir un paquete, atender el siguiente y **ver quién sigue** sin sacarlo.

```python
class ColaRecepcion:
    def __init__(self):
        self.cola = []          # ahora es un heap, no una lista común
        self.contador = 0       # desempate para prioridades iguales

    def recibir(self, paquete):
      paquete.orden = self.contador
      self.contador += 1
      heapq.heappush(self.cola, (paquete.prioridad, paquete.orden, paquete))

    def reinsertar(self, paquete):
      # Devuelve un paquete a la cola conservando su número de llegada original
      heapq.heappush(self.cola, (paquete.prioridad, paquete.orden, paquete))

    def atender_siguiente(self):
      if self.isVacia():
        return None
      return heapq.heappop(self.cola)[2]

    def consultar_siguiente(self):
      if self.isVacia():
        return None
      return self.cola[0][2]   # el mínimo del heap siempre está en el índice 0

    def cambiar_prioridad(self, paquete, nueva_prioridad):
      # Un heap no permite modificar un elemento: se quita y se vuelve a insertar
      self.cola = [entrada for entrada in self.cola if entrada[2] is not paquete]
      heapq.heapify(self.cola)
      paquete.prioridad = nueva_prioridad
      heapq.heappush(self.cola, (paquete.prioridad, paquete.orden, paquete))

    def cantidad(self):
      return len(self.cola)

    def isVacia(self):
      return len(self.cola) == 0
```

### Mínimo 3 · Consulta por código de seguimiento (tabla hash propia)
Implementen la tabla hash completa: función hash, insertar, buscar y manejo de colisiones. **Sin `dict` ni `set`.**

```python
# Tu código aquí: tabla hash

class TablaHash:
  def __init__(self, size=11, factor_max=0.75):
    self.size = size
    self.factor_max = factor_max
    self.cantidad = 0
    self.tabla = [[] for i in range(self.size)]  # llenar la tabla con los buckets

  def hash(self, key):
    # Polinómica: cada carácter pesa según su posición (a diferencia de solo sumar)
    h = 0
    for c in str(key):
      h = (h * 31 + ord(c)) % self.size
    return h

  def put(self, key, value):
    espacio = self.tabla[self.hash(key)]

    for par in espacio:
      if par[0] == key:
        par[1] = value      # la clave ya existía: se actualiza el valor
        return

    espacio.append([key, value])
    self.cantidad += 1
    if self.cantidad / self.size > self.factor_max:
      self._redimensionar()

  def _redimensionar(self):
    viejos = self.tabla
    self.size = self.size * 2
    self.tabla = [[] for i in range(self.size)]
    # La posición depende del tamaño, por eso hay que recalcularla para cada elemento
    for espacio in viejos:
      for clave, valor in espacio:
        self.tabla[self.hash(clave)].append([clave, valor])

  def buscar(self, key):
    # Devuelve (valor, comparaciones). Si no existe, valor = None
    comparaciones = 0
    for par in self.tabla[self.hash(key)]:
      comparaciones += 1
      if par[0] == key:
        return par[1], comparaciones
    return None, comparaciones

  def get(self, key):
    return self.buscar(key)[0]
```

### Mínimo 4 · Deshacer acciones
Decidan qué acciones se registran y **qué información hay que guardar** para poder revertirlas.

```python

class PilaDeshacer:
  # Pila (LIFO): lo último que se hizo es lo primero que se deshace.
  # Cada registro es [accion, paquete, dato_anterior]:
  #   'atender'   -> no necesita dato: al deshacer, el paquete vuelve a la cola y a 'espera'
  #   'entregar'  -> no necesita dato: al deshacer, el paquete vuelve a 'atendido'
  #   'prioridad' -> dato_anterior = la prioridad que tenía antes del cambio
  # (Recibir un paquete NO se registra: no se puede deshacer)
  def __init__(self):
    self.historial = []

  def registrar(self, accion, paquete, dato_anterior=None):
    self.historial.append([accion, paquete, dato_anterior])

  def sacar_ultimo(self):
    if len(self.historial) == 0:
      return None
    return self.historial.pop()
```

### Mínimo 5 · Búsqueda de zona por código (búsqueda binaria)
La lista de zonas debe estar **ordenada** antes de buscar.

```python
def ordenar_zonas(lista_zonas):
  # La lista de zonas es fija: se ordena una sola vez por código antes de buscar
  return sorted(lista_zonas, key=lambda zona: zona.codigo)

def busqueda_binaria_zonas(lista_zonas, codigo_buscado):
  # La lista debe estar ordenada por código. Devuelve (zona, comparaciones)
  inicio = 0
  fin = len(lista_zonas)-1
  comparaciones = 0

  while inicio <= fin:
    medio = (inicio+fin) // 2
    actual = lista_zonas[medio]
    comparaciones += 1          # una comparación por vuelta: ¿es el del medio?

    if actual.codigo == codigo_buscado:
      return actual, comparaciones
    elif actual.codigo < codigo_buscado:
      inicio = medio+1
    else:
      fin = medio-1

  return None, comparaciones
```

### Mínimo 6 · Búsqueda de paquetes por nombre de cliente (búsqueda secuencial)
Piensen qué devuelve su búsqueda si hay varios paquetes con el mismo nombre.

```python
def normalizar_nombre(nombre):
  # Ignora mayúsculas y espacios de más: ' juan   PEREZ ' -> 'juan perez'
  return ' '.join(nombre.split()).lower()

def busqueda_secuencial_cliente(lista_paquetes, cliente_buscado):
  # Devuelve (lista_de_paquetes_encontrados, comparaciones)
  # Si el cliente tiene varios paquetes (o hay clientes con el mismo nombre), los devuelve todos.
  # Por eso siempre recorre la lista completa.
  buscado = normalizar_nombre(cliente_buscado)
  encontrados = []
  comparaciones = 0

  for paquete in lista_paquetes:
    comparaciones += 1
    if normalizar_nombre(paquete.cliente.nombre) == buscado:
      encontrados.append(paquete)

  return encontrados, comparaciones
```

### Sistema integrador
Une todas las estructuras para que cada paquete quede registrado en todas a la vez y para que las acciones (atender, entregar, cambiar prioridad, deshacer) sean coherentes entre sí. Es el único que le habla al usuario.

```python
class CentroDistribucion:
  # Cada paquete vive en estas estructuras, siempre como el MISMO objeto (no copias):
  #   lista_general: TODOS los paquetes (la única que incluye los que no tienen código)
  #   tabla: solo los paquetes CON código, para buscarlos por código
  #   cola: los paquetes que siguen en espera de ser atendidos
  def __init__(self, zonas):
    self.zonas = ordenar_zonas(zonas)
    self.cola = ColaRecepcion()
    self.tabla = TablaHash()
    self.historial = PilaDeshacer()
    self.lista_general = []

  # Recepción y atención
  def recibir(self, paquete):
    if paquete.codigo is not None and self.tabla.get(paquete.codigo) is not None:
      print(f'Error: ya existe un paquete con el código {paquete.codigo}')
      return False
    self.lista_general.append(paquete)
    if paquete.codigo is not None:
      self.tabla.put(paquete.codigo, paquete)
    self.cola.recibir(paquete)
    nombre = paquete.codigo if paquete.codigo is not None else 'SIN CODIGO'
    print(f'Paquete {nombre} recibido y puesto en cola')
    return True

  def ver_siguiente(self):
    siguiente = self.cola.consultar_siguiente()
    if siguiente is None:
      print('No hay ningun paquete en la cola')
    return siguiente

  def atender(self):
    paquete = self.cola.atender_siguiente()
    if paquete is None:
      print('No hay ningun paquete en la cola')
      return None
    paquete.estado = 'atendido'
    self.historial.registrar('atender', paquete)
    return paquete

  def entregar(self, paquete):
    if paquete.estado != 'atendido':
      print(f'No se puede entregar: el paquete está en estado "{paquete.estado}" (debe estar "atendido")')
      return False
    paquete.estado = 'entregado'
    self.historial.registrar('entregar', paquete)
    return True

  def cambiar_prioridad(self, paquete, nueva_prioridad):
    if paquete.estado != 'espera':
      print('Solo se puede cambiar la prioridad de un paquete que sigue en espera')
      return False
    if nueva_prioridad not in (1, 2, 3):
      print('La prioridad debe ser 1 (alta), 2 (media) o 3 (baja)')
      return False
    anterior = paquete.prioridad
    self.cola.cambiar_prioridad(paquete, nueva_prioridad)
    self.historial.registrar('prioridad', paquete, anterior)
    return True

  # Deshacer
  def deshacer(self):
    registro = self.historial.sacar_ultimo()
    if registro is None:
      print('No hay acciones en el historial de cambios')
      return False
    accion, paquete, anterior = registro
    nombre = paquete.codigo if paquete.codigo is not None else 'SIN CODIGO'
    if accion == 'atender':
      paquete.estado = 'espera'
      self.cola.reinsertar(paquete)
      print(f'Se deshizo la atención del paquete {nombre}: volvió a la cola')
    elif accion == 'entregar':
      paquete.estado = 'atendido'
      print(f'Se deshizo la entrega del paquete {nombre}: volvió al estado atendido')
    elif accion == 'prioridad':
      self.cola.cambiar_prioridad(paquete, anterior)
      print(f'Se deshizo el cambio de prioridad del paquete {nombre}: volvió a prioridad {anterior}')
    return True

  # Búsquedas (informan si no encuentran)
  def buscar_por_codigo(self, codigo):
    paquete, comparaciones = self.tabla.buscar(codigo)
    if paquete is None:
      print(f'El paquete {codigo} no se encontró')
    return paquete

  def buscar_por_cliente(self, nombre):
    encontrados, comparaciones = busqueda_secuencial_cliente(self.lista_general, nombre)
    if len(encontrados) == 0:
      print(f'No se encontraron paquetes del cliente "{nombre}"')
    return encontrados

  def buscar_zona(self, codigo):
    zona, comparaciones = busqueda_binaria_zonas(self.zonas, codigo)
    if zona is None:
      print(f'La zona {codigo} no existe')
    return zona
```

### Mínimo 8 · Medición de comparaciones
Midan cuántas comparaciones hace cada búsqueda (secuencial, binaria y hash) con **100, 1000 y 10000** datos y muestren una tabla con los resultados. Después escriban en la celda siguiente qué observan.

```python
# Convención (ver registro de decisiones): una comparación = comparar la clave buscada con una clave guardada.
from random import seed, shuffle, randint

def busqueda_secuencial_codigo(lista_paquetes, codigo_buscado):
  # Igual que la secuencial por nombre, pero por código y SE DETIENE al encontrarlo.
  # (La de nombre siempre recorre todo porque devuelve todos los que coincide, no sirve para comparar.)
  comparaciones = 0
  for paquete in lista_paquetes:
    comparaciones += 1
    if paquete.codigo == codigo_buscado:
      return paquete, comparaciones
  return None, comparaciones

def promedio(buscar, claves):
  # Promedio de comparaciones de una búsqueda sobre muchas claves
  return sum(buscar(clave)[1] for clave in claves) / len(claves)

def medir(n, consultas=200):
  # Los 3 métodos buscan sobre los MISMOS n paquetes
  cliente = Cliente('Cliente de prueba', '0999999999')
  zona = Zona(1, 'Zona de prueba')
  numeros = list(range(n))
  shuffle(numeros)                                  # los paquetes llegan en desorden
  paquetes = [Paquete(f'ENV{100 + k}', cliente, zona, 'prueba') for k in numeros]

  tabla = TablaHash()
  for p in paquetes:
    tabla.put(p.codigo, p)
  ordenados = sorted(paquetes, key=lambda p: p.codigo)   # la binaria exige orden (ordenar no se cuenta)

  existentes = [paquetes[randint(0, n - 1)].codigo for _ in range(consultas)]
  inexistentes = [f'ENV{100 + n + k}' for k in range(consultas)]   # mismo formato, pero no están

  # Los tres métodos deben encontrar el paquete correcto (si no, el conteo no vale)
  for c in existentes:
    assert busqueda_secuencial_codigo(paquetes, c)[0].codigo == c
    assert busqueda_binaria_zonas(ordenados, c)[0].codigo == c   # sirve para cualquier lista ordenada con .codigo
    assert tabla.buscar(c)[0].codigo == c

  buscadores = [lambda c: busqueda_secuencial_codigo(paquetes, c),
                lambda c: busqueda_binaria_zonas(ordenados, c),
                lambda c: tabla.buscar(c)]
  fila = [promedio(b, existentes) for b in buscadores] + [promedio(b, inexistentes) for b in buscadores]
  mayor_casilla = max(len(casilla) for casilla in tabla.tabla)
  return n, fila, (tabla.size, tabla.cantidad / tabla.size, mayor_casilla)

seed(42)   # mismos datos en cada ejecución
resultados = [medir(n) for n in (100, 1000, 10000)]

print('Promedio de comparaciones por búsqueda (200 búsquedas por tamaño)\n')
print(f'{"":>7} | {"CÓDIGO EXISTENTE":^29} | {"CÓDIGO INEXISTENTE":^29}')
print(f'{"Datos":>7} | {"Secuencial":>10} {"Binaria":>9} {"Hash":>8} | {"Secuencial":>10} {"Binaria":>9} {"Hash":>8}')
for n, f, _ in resultados:
  print(f'{n:>7} | {f[0]:>10.1f} {f[1]:>9.1f} {f[2]:>8.2f} | {f[3]:>10.1f} {f[4]:>9.1f} {f[5]:>8.2f}')

print('\nEstado de la tabla hash al terminar')
print(f'{"Datos":>7} | {"Casillas":>8} | {"Elementos por casilla":>21} | {"Casilla más llena":>17}')
for n, _, (casillas, carga, mayor) in resultados:
  print(f'{n:>7} | {casillas:>8} | {carga:>21.2f} | {mayor:>17}')
```

### Análisis de las mediciones
¿Qué pasa con las comparaciones al multiplicar los datos por 10? ¿Cuál búsqueda crece más rápido? ¿Su tabla hash se comportó como esperaban?


*   **Secuencial:** crece al mismo ritmo que los datos. Con 10 veces más paquetes hace unas 10 veces más comparaciones (en promedio, la mitad de N; si el código no existe, N completo). Es la que crece más rápido.
*   **Binaria:** crece muy despacio. Cada vez que los datos se multiplican por 10 solo suma unas 3 o 4 comparaciones (log₂ N). Pero exige que la lista esté ordenada, y ordenar tiene un costo que esta medición no cuenta.

*   **Hash:** casi no crece: se queda en alrededor de 1 comparación por búsqueda sin importar si hay 100 o 10.000 paquetes. Sí se comportó como esperábamos, y la tabla de estado lo explica: al duplicar su tamaño cuando se llena, mantiene menos de 1 elemento por casilla y ninguna casilla acumula muchos.

*  **Para el dueño:** con 1 millón de paquetes la secuencial haría cientos de miles de comparaciones, la binaria unas 20 y la hash seguiría en alrededor de 1.
