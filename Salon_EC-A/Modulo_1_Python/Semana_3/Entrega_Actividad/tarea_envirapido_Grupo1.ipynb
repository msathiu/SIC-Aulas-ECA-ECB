### Registro de decisiones de diseño
Qué decidimos para cada punto de libertad y **por qué** (alimenta el informe):

- **Paquetes urgentes:** *no rompen el orden de llegada dentro de su categoría*. Hay **dos colas FIFO**: urgentes y normales. Los urgentes se atienden primero, **pero** después de 5 urgentes seguidos se atiende 1 normal (regla anti-inanición). *Por qué:* un urgente debe salir rápido, pero si llegan urgentes sin parar, los paquetes normales nunca se entregarían y el dueño recibiría quejas. Dentro de cada cola nadie se salta a nadie.
- **Nombres de cliente repetidos:** la búsqueda por nombre **devuelve una lista con todas las coincidencias** (nunca solo la primera). Los nombres se **normalizan** (sin tildes, mayúsculas/minúsculas ni espacios extra) para que "María  Pérez" y "maria perez" coincidan. El empleado distingue a los homónimos con el teléfono y el código. *Por qué:* devolver solo el primero entregaría el paquete a la persona equivocada.
- **Tamaño y función hash / tabla llena:** tabla de **101 posiciones (número primo)** con **encadenamiento separado** (listas enlazadas propias) y función **djb2** (`h = h*33 + ord(c)`) módulo el tamaño. Cuando el **factor de carga supera 0.75** la tabla **se redimensiona**: duplica su tamaño al siguiente primo y reubica todos los nodos (*rehash*). *Por qué:* el encadenamiento nunca se "llena" (no falla), y redimensionar mantiene las cadenas cortas (≈1-2 comparaciones) sin importar cuántos paquetes lleguen. La medición del Mínimo 8 lo demuestra.
- **Qué se puede deshacer y qué no:** se deshacen las acciones que **cambian el estado de los paquetes**: *recibir*, *atender (marcar entregado)* y *reasignar zona*, en orden inverso (pila, LIFO), con un límite de **100 acciones**. **No** se deshacen: las consultas/búsquedas (no cambian nada) y las altas de zonas (son cambios administrativos del catálogo). *Por qué:* como se deshace siempre en orden inverso, el sistema vuelve exactamente al estado anterior, y por eso deshacer "recibir" siempre encuentra el paquete al final de su cola (operación O(1)).
- **Zonas (¿cambian? ¿cómo se mantienen ordenadas?):** sí pueden cambiar. Cada alta usa **búsqueda binaria para hallar la posición de inserción** y se inserta ahí, así la lista **siempre está ordenada** (nunca se reordena toda). Un código de zona repetido se rechaza. *Por qué:* ordenar de nuevo en cada alta costaría O(n log n); insertar en la posición correcta cuesta O(log n) para encontrarla + el desplazamiento.
- **Cómo se informa que algo no se encontró:** ninguna función "revienta" ni devuelve `None` pelado al usuario: todas las operaciones del sistema devuelven un objeto **`Resultado(ok, mensaje, dato)`** con un mensaje claro y accionable (p. ej. *"No existe ningún paquete con el código X. Verifique el código o búsquelo por nombre del cliente"*). Las estructuras internas devuelven `None`/`-1`/lista vacía y el sistema lo traduce a mensajes. *Por qué:* el dueño y sus empleados no saben de excepciones; necesitan saber qué hacer después.

### Preparación: módulos
Importen aquí lo que necesiten (por ejemplo `randint` para generar datos de prueba). **Recuerden:** no usar `dict` ni `set` para la tabla hash principal.

```python
import unicodedata
import math
from random import randint, sample, choice, seed  # datos de prueba reproducibles

# --- Utilidades pequeñas (no son estructuras de datos del problema) ---------------

def normalizar(texto):
    """Minúsculas, sin tildes y sin espacios repetidos: 'María  PÉREZ ' -> 'maria perez'."""
    sin_tildes = "".join(c for c in unicodedata.normalize("NFD", texto)
                         if unicodedata.category(c) != "Mn")
    return " ".join(sin_tildes.casefold().split())


def es_primo(n):
    if n < 2:
        return False
    for d in range(2, int(math.isqrt(n)) + 1):
        if n % d == 0:
            return False
    return True


def siguiente_primo(n):
    """Menor primo >= n. Usamos tamaños primos en la tabla hash para repartir mejor las claves."""
    while not es_primo(n):
        n += 1
    return n
```

### Mínimo 1 · Modelado: paquete, cliente y zona
Definan las clases o estructuras. Antes de programar, escriban en papel los **sustantivos** y los **verbos** del problema.

```python
# ======================= MÍNIMO 1 · MODELADO =======================
PENDIENTE = "PENDIENTE"    # está en el centro, esperando ser atendido
ENTREGADO = "ENTREGADO"    # ya fue atendido/entregado


class Cliente:
    """Quién envía/recibe. Guarda el nombre tal cual y una versión normalizada para buscar."""
    def __init__(self, nombre, telefono=""):
        if not nombre or not nombre.strip():
            raise ValueError("El cliente necesita un nombre")
        self.nombre = " ".join(nombre.split())
        self.telefono = telefono
        self.nombre_clave = normalizar(self.nombre)   # clave de búsqueda por nombre

    def __repr__(self):
        return f"Cliente({self.nombre!r}, tel={self.telefono!r})"


class Zona:
    """Zona/sucursal de reparto, identificada por un código numérico único."""
    def __init__(self, codigo, nombre):
        self.codigo = int(codigo)
        self.nombre = nombre

    def __repr__(self):
        return f"Zona({self.codigo}, {self.nombre!r})"


class Paquete:
    """Un paquete: código de seguimiento, su cliente, su zona, prioridad y estado."""
    def __init__(self, codigo, cliente, zona, urgente=False, orden_llegada=0):
        self.codigo = codigo.strip().upper()      # clave única (normalizada)
        self.cliente = cliente                    # objeto Cliente
        self.zona = zona                          # objeto Zona
        self.urgente = urgente
        self.estado = PENDIENTE
        self.orden_llegada = orden_llegada        # nº de llegada: demuestra el orden

    def __repr__(self):
        prio = "URGENTE" if self.urgente else "normal"
        return (f"Paquete({self.codigo} | {self.cliente.nombre} | zona {self.zona.codigo} | "
                f"{prio} | {self.estado})")


class Resultado:
    """Respuesta uniforme del sistema al usuario: éxito/fracaso + mensaje + dato opcional."""
    def __init__(self, ok, mensaje, dato=None):
        self.ok, self.mensaje, self.dato = ok, mensaje, dato

    def __bool__(self):
        return self.ok

    def __repr__(self):
        return ("✔ " if self.ok else "✘ ") + self.mensaje
```

### Mínimo 2 · Recepción y atención en orden de llegada
Necesitan: recibir un paquete, atender el siguiente y **ver quién sigue** sin sacarlo.

```python
# ======================= MÍNIMO 2 · COLA EN ORDEN DE LLEGADA =======================
# Hecha por nosotros con una lista DOBLEMENTE enlazada: encolar, desencolar y ver el frente
# cuestan O(1). (list.pop(0) de Python costaría O(n): mover todos los elementos.)

class NodoDoble:
    __slots__ = ("dato", "anterior", "siguiente")

    def __init__(self, dato):
        self.dato = dato
        self.anterior = None
        self.siguiente = None


class ColaDoble:
    """Cola FIFO. Además permite operar por el otro extremo, lo que necesita el 'deshacer'."""
    def __init__(self):
        self.frente = None
        self.final = None
        self.tamano = 0

    def __len__(self):
        return self.tamano

    def esta_vacia(self):
        return self.tamano == 0

    def encolar(self, dato):                 # entra por el final
        nodo = NodoDoble(dato)
        if self.final is None:
            self.frente = self.final = nodo
        else:
            nodo.anterior = self.final
            self.final.siguiente = nodo
            self.final = nodo
        self.tamano += 1

    def encolar_al_frente(self, dato):       # solo para deshacer una atención
        nodo = NodoDoble(dato)
        if self.frente is None:
            self.frente = self.final = nodo
        else:
            nodo.siguiente = self.frente
            self.frente.anterior = nodo
            self.frente = nodo
        self.tamano += 1

    def ver_frente(self):
        return None if self.frente is None else self.frente.dato

    def ver_final(self):
        return None if self.final is None else self.final.dato

    def desencolar(self):                    # sale por el frente
        if self.frente is None:
            return None
        nodo = self.frente
        self.frente = nodo.siguiente
        if self.frente is None:
            self.final = None
        else:
            self.frente.anterior = None
        self.tamano -= 1
        return nodo.dato

    def quitar_final(self):                  # solo para deshacer una recepción
        if self.final is None:
            return None
        nodo = self.final
        self.final = nodo.anterior
        if self.final is None:
            self.frente = None
        else:
            self.final.siguiente = None
        self.tamano -= 1
        return nodo.dato


class ColaAtencion:
    """Dos colas FIFO (urgentes / normales) con regla anti-inanición:
    los urgentes van primero, pero tras `limite_urgentes` urgentes seguidos sale 1 normal."""
    def __init__(self, limite_urgentes=5):
        self.urgentes = ColaDoble()
        self.normales = ColaDoble()
        self.limite_urgentes = limite_urgentes
        self.urgentes_seguidos = 0

    def __len__(self):
        return len(self.urgentes) + len(self.normales)

    def esta_vacia(self):
        return len(self) == 0

    def recibir(self, paquete):
        (self.urgentes if paquete.urgente else self.normales).encolar(paquete)

    def _cola_que_sigue(self):
        hay_u = not self.urgentes.esta_vacia()
        hay_n = not self.normales.esta_vacia()
        if hay_u and (not hay_n or self.urgentes_seguidos < self.limite_urgentes):
            return self.urgentes
        if hay_n:
            return self.normales
        return None

    def ver_siguiente(self):
        """¿Quién sigue? Lo muestra SIN sacarlo."""
        cola = self._cola_que_sigue()
        return None if cola is None else cola.ver_frente()

    def proximos(self, n):
        """Los próximos n paquetes en el orden en que serían atendidos (sin sacarlos)."""
        resultado, u, nor, seguidos = [], self.urgentes.frente, self.normales.frente, self.urgentes_seguidos
        while len(resultado) < n and (u is not None or nor is not None):
            if u is not None and (nor is None or seguidos < self.limite_urgentes):
                resultado.append(u.dato); u = u.siguiente; seguidos += 1
            else:
                resultado.append(nor.dato); nor = nor.siguiente; seguidos = 0
        return resultado

    def atender(self):
        """Saca al siguiente. Devuelve (paquete, urgentes_seguidos_antes) o (None, None) si no hay nadie."""
        cola = self._cola_que_sigue()
        if cola is None:
            return None, None
        seguidos_antes = self.urgentes_seguidos
        paquete = cola.desencolar()
        self.urgentes_seguidos = seguidos_antes + 1 if cola is self.urgentes else 0
        return paquete, seguidos_antes

    # --- operaciones inversas, solo las usa deshacer ---
    def devolver_al_frente(self, paquete, seguidos_antes):
        (self.urgentes if paquete.urgente else self.normales).encolar_al_frente(paquete)
        self.urgentes_seguidos = seguidos_antes

    def quitar_ultimo_recibido(self, paquete):
        cola = self.urgentes if paquete.urgente else self.normales
        if cola.ver_final() is paquete:
            cola.quitar_final()
            return True
        return False
```

### Mínimo 3 · Consulta por código de seguimiento (tabla hash propia)
Implementen la tabla hash completa: función hash, insertar, buscar y manejo de colisiones. **Sin `dict` ni `set`.**

```python
# ======================= MÍNIMO 3 · TABLA HASH PROPIA =======================
# Sin dict ni set. Cubetas = arreglo de listas enlazadas (encadenamiento separado).
# Una "comparación" = comparar la clave buscada con la clave de un nodo.

class NodoHash:
    __slots__ = ("clave", "valor", "siguiente")

    def __init__(self, clave, valor):
        self.clave = clave
        self.valor = valor
        self.siguiente = None


class TablaHash:
    FACTOR_MAXIMO = 0.75      # si cantidad/tamaño supera esto, la tabla crece

    def __init__(self, tamano_inicial=101, autoredimensionar=True):
        self.tamano = siguiente_primo(tamano_inicial)
        self.cubetas = [None] * self.tamano
        self.cantidad = 0
        self.autoredimensionar = autoredimensionar
        # estadísticas para el análisis
        self.colisiones = 0              # inserciones que cayeron en una cubeta ya ocupada
        self.redimensionamientos = 0
        self.comparaciones = 0           # comparaciones de la ÚLTIMA operación

    def __len__(self):
        return self.cantidad

    @staticmethod
    def _hash(clave):
        """djb2: h = h*33 + código del carácter (limitado a 32 bits)."""
        h = 5381
        for caracter in clave:
            h = (h * 33 + ord(caracter)) & 0xFFFFFFFF
        return h

    def _indice(self, clave):
        return self._hash(clave) % self.tamano

    def factor_carga(self):
        return self.cantidad / self.tamano

    def insertar(self, clave, valor):
        """Inserta. Devuelve False si la clave ya existía (no se duplican códigos)."""
        i = self._indice(clave)
        nodo, comps = self.cubetas[i], 0
        while nodo is not None:
            comps += 1
            if nodo.clave == clave:
                self.comparaciones = comps
                return False
            nodo = nodo.siguiente
        self.comparaciones = comps
        if self.cubetas[i] is not None:
            self.colisiones += 1                      # ¡colisión! se encadena en la misma cubeta
        nuevo = NodoHash(clave, valor)
        nuevo.siguiente = self.cubetas[i]             # se inserta al inicio de la cadena: O(1)
        self.cubetas[i] = nuevo
        self.cantidad += 1
        if self.autoredimensionar and self.factor_carga() > self.FACTOR_MAXIMO:
            self._redimensionar()
        return True

    def buscar(self, clave):
        """Devuelve el valor o None si no existe."""
        nodo, comps = self.cubetas[self._indice(clave)], 0
        while nodo is not None:
            comps += 1
            if nodo.clave == clave:
                self.comparaciones = comps
                return nodo.valor
            nodo = nodo.siguiente
        self.comparaciones = comps
        return None

    def eliminar(self, clave):
        """Quita la clave y devuelve su valor (None si no estaba). Lo usa 'deshacer recibir'."""
        i = self._indice(clave)
        nodo, previo, comps = self.cubetas[i], None, 0
        while nodo is not None:
            comps += 1
            if nodo.clave == clave:
                if previo is None:
                    self.cubetas[i] = nodo.siguiente
                else:
                    previo.siguiente = nodo.siguiente
                self.cantidad -= 1
                self.comparaciones = comps
                return nodo.valor
            previo, nodo = nodo, nodo.siguiente
        self.comparaciones = comps
        return None

    def _redimensionar(self):
        """Duplica el tamaño (al siguiente primo) y reubica TODOS los nodos con el nuevo módulo."""
        viejas = self.cubetas
        self.tamano = siguiente_primo(self.tamano * 2)
        self.cubetas = [None] * self.tamano
        for cabeza in viejas:
            nodo = cabeza
            while nodo is not None:
                siguiente = nodo.siguiente
                i = self._indice(nodo.clave)
                nodo.siguiente = self.cubetas[i]
                self.cubetas[i] = nodo
                nodo = siguiente
        self.redimensionamientos += 1

    def cadena_mas_larga(self):
        mayor = 0
        for cabeza in self.cubetas:
            largo, nodo = 0, cabeza
            while nodo is not None:
                largo += 1
                nodo = nodo.siguiente
            mayor = max(mayor, largo)
        return mayor
```

### Mínimo 4 · Deshacer acciones
Decidan qué acciones se registran y **qué información hay que guardar** para poder revertirlas.

```python
# ======================= MÍNIMO 4 · DESHACER (PILA) =======================
# Estructura: PILA (LIFO). Lo último que se hizo es lo primero que se deshace.
# Para revertir cada acción guardamos lo mínimo necesario:
#   RECIBIR   -> el paquete                          (revertir = quitarlo de cola, tabla y registro)
#   ATENDER   -> el paquete + urgentes_seguidos antes (revertir = volver al frente y estado PENDIENTE)
#   REASIGNAR -> el paquete + la zona anterior        (revertir = devolver la zona anterior)

RECIBIR, ATENDER, REASIGNAR = "RECIBIR", "ATENDER", "REASIGNAR"


class Accion:
    def __init__(self, tipo, paquete, dato_previo=None):
        self.tipo = tipo
        self.paquete = paquete
        self.dato_previo = dato_previo

    def __repr__(self):
        return f"Accion({self.tipo}, {self.paquete.codigo})"


class Historial:
    """Pila acotada: guarda las últimas `capacidad` acciones (descarta la más antigua)."""
    def __init__(self, capacidad=100):
        self.capacidad = capacidad
        self._pila = []
        self.descartadas = 0

    def __len__(self):
        return len(self._pila)

    def registrar(self, accion):               # push
        self._pila.append(accion)
        if len(self._pila) > self.capacidad:
            del self._pila[0]
            self.descartadas += 1

    def sacar(self):                           # pop (None si no hay historial)
        return self._pila.pop() if self._pila else None

    def ver_ultima(self):                      # peek
        return self._pila[-1] if self._pila else None
```

### Mínimo 5 · Búsqueda de zona por código (búsqueda binaria)
La lista de zonas debe estar **ordenada** antes de buscar.

```python
# ======================= MÍNIMO 5 · ZONAS ORDENADAS + BÚSQUEDA BINARIA =======================
# Una "comparación" = una sonda del arreglo (consultar un elemento del medio).

def busqueda_binaria(lista, clave, obtener_clave=lambda x: x):
    """Devuelve (indice, comparaciones). indice = -1 si no está. Requiere lista ORDENADA."""
    izq, der, comps = 0, len(lista) - 1, 0
    while izq <= der:
        medio = (izq + der) // 2
        comps += 1
        valor = obtener_clave(lista[medio])
        if valor == clave:
            return medio, comps
        if valor < clave:
            izq = medio + 1
        else:
            der = medio - 1
    return -1, comps


def posicion_insercion(lista, clave, obtener_clave=lambda x: x):
    """Variante de la binaria: primera posición donde `clave` cabe sin romper el orden."""
    izq, der = 0, len(lista)
    while izq < der:
        medio = (izq + der) // 2
        if obtener_clave(lista[medio]) < clave:
            izq = medio + 1
        else:
            der = medio
    return izq


class ListaZonas:
    """Lista de zonas SIEMPRE ordenada por código. Las altas se insertan en su lugar."""
    def __init__(self):
        self._zonas = []
        self.comparaciones = 0           # de la última búsqueda

    def __len__(self):
        return len(self._zonas)

    def __iter__(self):
        return iter(self._zonas)

    def codigos(self):
        return [z.codigo for z in self._zonas]

    def agregar(self, zona):
        """Inserta manteniendo el orden. Devuelve False si el código ya existe."""
        pos = posicion_insercion(self._zonas, zona.codigo, lambda z: z.codigo)
        if pos < len(self._zonas) and self._zonas[pos].codigo == zona.codigo:
            return False
        self._zonas.insert(pos, zona)
        return True

    def buscar(self, codigo):
        """Búsqueda binaria por código. Devuelve la Zona o None."""
        indice, self.comparaciones = busqueda_binaria(self._zonas, codigo, lambda z: z.codigo)
        return None if indice == -1 else self._zonas[indice]
```

### Mínimo 6 · Búsqueda de paquetes por nombre de cliente (búsqueda secuencial)
Piensen qué devuelve su búsqueda si hay varios paquetes con el mismo nombre.

```python
# ======================= MÍNIMO 6 · BÚSQUEDA SECUENCIAL POR NOMBRE =======================
# Una "comparación" = examinar un elemento de la lista.

def busqueda_secuencial(lista, clave, obtener_clave=lambda x: x):
    """Primer elemento cuya clave coincide. Devuelve (indice, comparaciones); -1 si no está."""
    comps = 0
    for i, elemento in enumerate(lista):
        comps += 1
        if obtener_clave(elemento) == clave:
            return i, comps
    return -1, comps


def busqueda_secuencial_nombre(paquetes, nombre, parcial=False):
    """Busca paquetes por nombre de cliente cuando NO hay código de seguimiento.
    - Devuelve (lista_de_coincidencias, comparaciones): TODAS las coincidencias, porque pueden
      existir varios clientes (o varios paquetes) con el mismo nombre.
    - Ignora tildes, mayúsculas y espacios repetidos. Con parcial=True basta con que el nombre
      CONTENGA el texto (p. ej. 'torres' encuentra 'Juan Torres').
    - Debe recorrer TODA la lista (no puede parar en la primera) -> O(n)."""
    clave = normalizar(nombre)
    coincidencias, comps = [], 0
    for paquete in paquetes:
        comps += 1
        nombre_paquete = paquete.cliente.nombre_clave
        if (clave in nombre_paquete) if parcial else (nombre_paquete == clave):
            coincidencias.append(paquete)
    return coincidencias, comps
```

### Integración · el sistema completo `CentroDistribucion`
Junta todas las estructuras: **cola** (orden de atención) + **tabla hash** (consulta por código) + **lista de zonas ordenada** (búsqueda binaria) + **registro** (búsqueda secuencial por nombre) + **pila** (deshacer). Cada operación devuelve un `Resultado` con mensaje claro.

```python
class CentroDistribucion:
    def __init__(self, tamano_hash=101, capacidad_historial=100, limite_urgentes=5):
        self.zonas = ListaZonas()                   # Mínimo 5
        self.cola = ColaAtencion(limite_urgentes)   # Mínimo 2
        self.indice = TablaHash(tamano_hash)        # Mínimo 3: código -> Paquete
        self.registro = []                          # todos los paquetes en orden de llegada (Mínimo 6)
        self.historial = Historial(capacidad_historial)   # Mínimo 4
        self._llegadas = 0
        self.total_recibidos = self.total_entregados = self.total_deshechos = 0

    # ---------- zonas ----------
    def agregar_zona(self, codigo, nombre):
        try:
            zona = Zona(codigo, nombre)
        except (ValueError, TypeError):
            return Resultado(False, f"Código de zona inválido: {codigo!r} (debe ser un número)")
        if not self.zonas.agregar(zona):
            return Resultado(False, f"La zona {zona.codigo} ya existe")
        return Resultado(True, f"Zona {zona.codigo} ({nombre}) agregada", zona)

    def buscar_zona(self, codigo):
        try:
            zona = self.zonas.buscar(int(codigo))
        except (ValueError, TypeError):
            return Resultado(False, f"Código de zona inválido: {codigo!r}")
        if zona is None:
            return Resultado(False, f"No existe la zona {codigo}. Zonas disponibles: {len(self.zonas)}")
        return Resultado(True, f"Zona {zona.codigo}: {zona.nombre}", zona)

    # ---------- recepción y atención ----------
    def recibir_paquete(self, codigo, nombre_cliente, codigo_zona, telefono="", urgente=False):
        codigo = (codigo or "").strip().upper()
        if not codigo:
            return Resultado(False, "El paquete necesita un código de seguimiento")
        zona = self.buscar_zona(codigo_zona)
        if not zona:
            return Resultado(False, f"No se puede recibir {codigo}: {zona.mensaje}")
        if self.indice.buscar(codigo) is not None:
            return Resultado(False, f"Ya existe un paquete con el código {codigo}")
        try:
            cliente = Cliente(nombre_cliente, telefono)
        except ValueError as error:
            return Resultado(False, f"No se puede recibir {codigo}: {error}")
        self._llegadas += 1
        paquete = Paquete(codigo, cliente, zona.dato, urgente, self._llegadas)
        self.indice.insertar(codigo, paquete)
        self.cola.recibir(paquete)
        self.registro.append(paquete)
        self.historial.registrar(Accion(RECIBIR, paquete))
        self.total_recibidos += 1
        tipo = "URGENTE" if urgente else "normal"
        return Resultado(True, f"Paquete {codigo} recibido ({tipo}) para {cliente.nombre}, zona {zona.dato.codigo}", paquete)

    def ver_siguiente(self):
        paquete = self.cola.ver_siguiente()
        if paquete is None:
            return Resultado(False, "No hay paquetes en espera")
        return Resultado(True, f"Sigue: {paquete}", paquete)

    def atender_siguiente(self):
        paquete, seguidos_antes = self.cola.atender()
        if paquete is None:
            return Resultado(False, "No hay paquetes en espera: la cola está vacía")
        paquete.estado = ENTREGADO
        self.total_entregados += 1
        self.historial.registrar(Accion(ATENDER, paquete, seguidos_antes))
        return Resultado(True, f"Atendido y marcado ENTREGADO: {paquete.codigo} ({paquete.cliente.nombre})", paquete)

    # ---------- consultas ----------
    def consultar_codigo(self, codigo):
        """'¿Dónde está mi paquete?' -> tabla hash, O(1) promedio."""
        codigo = (codigo or "").strip().upper()
        paquete = self.indice.buscar(codigo)
        if paquete is None:
            return Resultado(False, f"No existe ningún paquete con el código {codigo}. "
                                    "Verifique el código o búsquelo por nombre del cliente")
        return Resultado(True, f"{paquete.codigo}: {paquete.estado} (cliente {paquete.cliente.nombre}, "
                               f"zona {paquete.zona.codigo} {paquete.zona.nombre})", paquete)

    def buscar_por_cliente(self, nombre, parcial=False):
        """Paquetes sin código -> búsqueda secuencial por nombre (todas las coincidencias)."""
        encontrados, comps = busqueda_secuencial_nombre(self.registro, nombre, parcial)
        if not encontrados:
            return Resultado(False, f"No hay paquetes a nombre de '{nombre}'. Pruebe con búsqueda parcial (parcial=True)", [])
        return Resultado(True, f"{len(encontrados)} paquete(s) a nombre de '{nombre}' ({comps} comparaciones)", encontrados)

    # ---------- extra: reasignar ----------
    def reasignar_zona(self, codigo, codigo_zona_nueva):
        paquete = self.indice.buscar((codigo or "").strip().upper())
        if paquete is None:
            return Resultado(False, f"No existe el paquete {codigo}")
        if paquete.estado == ENTREGADO:
            return Resultado(False, f"{paquete.codigo} ya fue entregado: no se puede reasignar")
        zona = self.buscar_zona(codigo_zona_nueva)
        if not zona:
            return Resultado(False, f"No se reasignó: {zona.mensaje}")
        self.historial.registrar(Accion(REASIGNAR, paquete, paquete.zona))
        anterior, paquete.zona = paquete.zona, zona.dato
        return Resultado(True, f"{paquete.codigo} reasignado de zona {anterior.codigo} a {zona.dato.codigo}", paquete)

    # ---------- deshacer ----------
    def deshacer(self):
        accion = self.historial.sacar()
        if accion is None:
            return Resultado(False, "No hay acciones para deshacer")
        p = accion.paquete
        if accion.tipo == ATENDER:
            p.estado = PENDIENTE
            self.cola.devolver_al_frente(p, accion.dato_previo)
            self.total_entregados -= 1
            mensaje = f"Deshecho: {p.codigo} vuelve al frente de la cola como PENDIENTE"
        elif accion.tipo == RECIBIR:
            if not self.cola.quitar_ultimo_recibido(p):
                raise RuntimeError("Inconsistencia: el paquete recibido no estaba al final de su cola")
            self.indice.eliminar(p.codigo)
            self.registro.pop()
            self.total_recibidos -= 1
            self._llegadas -= 1
            mensaje = f"Deshecho: se anuló la recepción de {p.codigo}"
        else:  # REASIGNAR
            p.zona = accion.dato_previo
            mensaje = f"Deshecho: {p.codigo} vuelve a la zona {p.zona.codigo}"
        self.total_deshechos += 1
        return Resultado(True, mensaje, p)

    # ---------- extra: estadísticas del día ----------
    def reporte(self):
        t = self.indice
        return (f"Recibidos: {self.total_recibidos} | Entregados: {self.total_entregados} | "
                f"En espera: {len(self.cola)} (urgentes: {len(self.cola.urgentes)}) | "
                f"Acciones deshechas: {self.total_deshechos}\n"
                f"Tabla hash: tamaño {t.tamano}, factor de carga {t.factor_carga():.2f}, "
                f"colisiones {t.colisiones}, cadena más larga {t.cadena_mas_larga()}, "
                f"redimensionamientos {t.redimensionamientos}")
```

### Mínimo 8 · Medición de comparaciones
Midan cuántas comparaciones hace cada búsqueda (secuencial, binaria y hash) con **100, 1000 y 10000** datos y muestren una tabla con los resultados. Después escriban en la celda siguiente qué observan.

```python
# ======================= MÍNIMO 8 · MEDICIÓN DE COMPARACIONES =======================
# Misma definición para las tres: 1 comparación = mirar un elemento/nodo y compararlo con la clave.
# Se buscan 500 códigos que SÍ existen (éxito) y 500 que NO existen (fracaso), al azar.

seed(2026)

def generar_paquetes(n):
    cliente, zona = Cliente("Cliente de prueba"), Zona(1, "Zona")
    return [Paquete("EV-%08d" % c, cliente, zona) for c in sample(range(10_000_000), n)]   # códigos únicos, desordenados


def medir(n, muestra=500, autoredimensionar=True):
    paquetes = generar_paquetes(n)                                    # orden de llegada (desordenado)
    tabla = TablaHash(101, autoredimensionar)
    for p in paquetes:
        tabla.insertar(p.codigo, p)
    ordenados = sorted(paquetes, key=lambda p: p.codigo)               # la binaria exige lista ordenada

    existentes = [choice(paquetes).codigo for _ in range(muestra)]
    inexistentes = []
    while len(inexistentes) < muestra:
        c = "EV-%08d" % randint(0, 9_999_999)
        if tabla.buscar(c) is None:
            inexistentes.append(c)

    def estadisticas(claves):
        sec = [busqueda_secuencial(paquetes, c, lambda p: p.codigo)[1] for c in claves]
        bina = [busqueda_binaria(ordenados, c, lambda p: p.codigo)[1] for c in claves]
        has = []
        for c in claves:
            tabla.buscar(c)
            has.append(tabla.comparaciones)
        return [(sum(v) / len(v), max(v)) for v in (sec, bina, has)]

    return estadisticas(existentes), estadisticas(inexistentes), tabla


TAMANOS = [100, 1000, 10000]
resultados = [(n,) + medir(n) for n in TAMANOS]

def imprimir(titulo_tabla, indice):
    print(f"\n{titulo_tabla}")
    print(f"{'N':>7} | {'Secuencial':^18} | {'Binaria':^18} | {'Hash propia':^18}")
    print(f"{'':>7} | {'promedio':>8} {'máx':>8} | {'promedio':>8} {'máx':>8} | {'promedio':>8} {'máx':>8}")
    print("-" * 7 + "-+-" + "-+-".join(["-" * 18] * 3))
    for n, ok, fallo, _ in resultados:
        fila = (ok, fallo)[indice]
        print(f"{n:>7} | " + " | ".join(f"{a:>8.1f} {m:>8}" for a, m in fila))

imprimir("Búsquedas EXITOSAS (el código existe) — comparaciones", 0)
imprimir("Búsquedas FALLIDAS (el código NO existe) — comparaciones", 1)

print("\nCrecimiento al multiplicar los datos por 10 (promedio, búsquedas exitosas):")
for i in range(1, len(resultados)):
    ant, act = resultados[i - 1][1], resultados[i][1]
    print(f"  {resultados[i-1][0]} -> {resultados[i][0]}:  secuencial x{act[0][0]/ant[0][0]:.1f} | "
          f"binaria +{act[1][0]-ant[1][0]:.1f} comparaciones | hash x{act[2][0]/ant[2][0]:.2f}")

print("\nEstado de la tabla hash tras insertar N claves (empezando en 101 posiciones):")
print(f"{'N':>7} | {'tamaño':>7} | {'factor carga':>12} | {'cadena máx':>10} | {'colisiones':>10} | {'rehash':>6}")
for n, _, _, t in resultados:
    print(f"{n:>7} | {t.tamano:>7} | {t.factor_carga():>12.2f} | {t.cadena_mas_larga():>10} | {t.colisiones:>10} | {t.redimensionamientos:>6}")

print("\nTeoría: secuencial ≈ N/2 (éxito) y N (fallo) | binaria ≈ log2(N) | hash ≈ 1 + α/2 (éxito) y α (fallo), α = factor de carga")
print("log2(N) para cada N:", [round(math.log2(n), 1) for n in TAMANOS])

# ¿Y si NO redimensionáramos la tabla? (experimento para defender nuestra decisión)
print("\nExperimento: tabla hash de tamaño FIJO 101 (sin redimensionar) — búsquedas exitosas")
print(f"{'N':>7} | {'hash fija prom':>14} | {'hash propia prom':>16}")
for n in TAMANOS:
    fija = medir(n, autoredimensionar=False)[0][2][0]
    nuestra = [r for r in resultados if r[0] == n][0][1][2][0]
    print(f"{n:>7} | {fija:>14.1f} | {nuestra:>16.2f}")

# Gráfico (opcional): escala logarítmica para ver el crecimiento de cada búsqueda
try:
    import matplotlib.pyplot as plt
    plt.figure(figsize=(6, 4))
    for etiqueta, k in (("Secuencial", 0), ("Binaria", 1), ("Hash propia", 2)):
        plt.plot(TAMANOS, [r[1][k][0] for r in resultados], marker="o", label=etiqueta)
    plt.xscale("log"); plt.yscale("log")
    plt.xlabel("N (cantidad de paquetes)"); plt.ylabel("comparaciones promedio")
    plt.title("Comparaciones por búsqueda (escala log-log)"); plt.legend(); plt.grid(True, which="both", alpha=.3)
    plt.tight_layout(); plt.show()
except ImportError:
    print("(matplotlib no está instalado: el gráfico es opcional)")
```

### Análisis de las mediciones
*(Números de nuestra ejecución con semilla fija; los valores exactos pueden variar un poco al cambiar la semilla, pero las conclusiones no.)*

**¿Qué pasa con las comparaciones al multiplicar los datos por 10?**

| Búsqueda | 100 datos | 1 000 datos | 10 000 datos | Comportamiento |
|---|---|---|---|---|
| Secuencial | ≈ 50 | ≈ 493 | ≈ 5 100 | se **multiplica ×10** cada vez → crecimiento lineal O(n) |
| Binaria | ≈ 5.8 | ≈ 9.0 | ≈ 12.5 | **suma ≈ 3.3** cada vez (log₂ 10 ≈ 3.32) → crecimiento logarítmico O(log n) |
| Hash propia | ≈ 1.3 | ≈ 1.3 | ≈ 1.3 | **no cambia** → O(1) promedio |

**¿Cuál búsqueda crece más rápido?** La secuencial, por mucho: con 10 000 paquetes hace unas **4 000 veces más comparaciones** que la hash y unas 400 veces más que la binaria. En las búsquedas fallidas es aún peor: debe revisar los N elementos (10 000 comparaciones), mientras que la hash apenas necesita ≈ 0.8 (la mayoría de las veces la cubeta está vacía).

**¿Nuestra tabla hash se comportó como esperábamos?** Sí. La teoría dice que con encadenamiento el promedio es ≈ 1 + α/2 comparaciones (α = factor de carga). Como la tabla se redimensiona cuando α > 0.75, α nunca pasó de ≈ 0.72 y el promedio se mantuvo en ≈ 1.3 sin importar N. La cadena más larga fue de solo 4-5 nodos con 10 000 claves, lo que indica que la función djb2 con tamaño primo reparte bien. Hubo colisiones (4 214 con 10 000 claves), pero el encadenamiento las resuelve sin perder ningún dato.

**Experimento clave (¿valió la pena redimensionar?):** con una tabla de tamaño **fijo (101)**, el promedio sube con N (≈ 1.6 → ≈ 6 → ≈ 50 comparaciones, porque las cadenas se alargan y la búsqueda vuelve a ser lineal, solo que dividida entre 101). Con redimensionamiento se queda en ≈ 1.3. Por eso lo incluimos en el diseño.

**Conclusión para el dueño:** para "¿dónde está mi paquete?" la tabla hash es la mejor opción (rápida sin importar cuántos paquetes haya). La binaria es excelente para las zonas (lista ordenada que casi no cambia). La secuencial solo se usa donde no hay alternativa (búsqueda por nombre, sin código), y es el único punto que se volvería lento con millones de paquetes: ahí la mejora sería un segundo índice hash por nombre.
