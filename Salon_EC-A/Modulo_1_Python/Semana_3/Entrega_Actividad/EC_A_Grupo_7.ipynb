### Mínimo 1 · Modelado: paquete, cliente y zona
Definan las clases o estructuras. Antes de programar, escriban en papel los **sustantivos** y los **verbos** del problema.

```python
# Tu código aquí: clases Paquete, Cliente, Zona

class Zona:
    def __init__(self, codigo_numerico, provincia, canton, parroquia, nombre_sucursal=""):
        self.codigo_numerico = int(codigo_numerico)
        self.provincia = provincia
        self.canton = canton
        self.parroquia = parroquia
        self.nombre_sucursal = nombre_sucursal if nombre_sucursal else f"Sucursal {canton} - {parroquia}"

    def getCodigoNumerico(self):
      return self.codigo_numerico

    def __str__(self):
        return f"[Zona {self.codigo_numerico}] {self.nombre_sucursal} ({self.provincia}, {self.canton}, {self.parroquia})"



class Persona:
    def __init__(self, nombre, apellido, cedula, telefono, edad=None):
        self.nombre = nombre.strip()
        self.apellido = apellido.strip()
        self.cedula = str(cedula).strip()
        self.telefono = str(telefono).strip()
        self.edad = edad

    def nombre_completo(self):
        return f"{self.nombre} {self.apellido}"

    def __str__(self):
        return f"{self.nombre_completo()} (C.I.: {self.cedula} | Tel: {self.telefono})"


class Cliente(Persona):
    def __init__(self, nombre, apellido, cedula, telefono, direccion, edad=None):
        super().__init__(nombre, apellido, cedula, telefono, edad)
        self.direccion = direccion

    def __str__(self):
        return f"Cliente: {self.nombre_completo()} | C.I.: {self.cedula} | Tel: {self.telefono} | Dir: {self.direccion}"


class Empleado(Persona):
    def __init__(self, nombre, apellido, cedula, telefono, codigo_empleado, cargo="Operador", sueldo=0.0, edad=None):
        super().__init__(nombre, apellido, cedula, telefono, edad)
        self.codigo_empleado = codigo_empleado
        self.cargo = cargo
        self.sueldo = sueldo

    def __str__(self):
        return f"Empleado [{self.codigo_empleado}]: {self.nombre_completo()} - Cargo: {self.cargo}"


class Paquete:
    def __init__(
        self,
        numero_llegada,
        duenio,
        codigo_zona,
        direccion_destino,
        codigo_seguimiento=None,
        es_urgente=False,
        estado="En recepción",
        fecha_llegada="",
        fecha_salida="",
    ):
        self.numero_llegada = numero_llegada
        self.duenio = duenio
        self.codigo_zona = int(codigo_zona)
        self.direccion_destino = direccion_destino
        self.codigo_seguimiento = codigo_seguimiento
        self.es_urgente = es_urgente
        self.estado = estado
        self.fecha_llegada = fecha_llegada
        self.fecha_salida = fecha_salida

    def getUrgencia(self):
      return self.es_urgente

    def getDuenio(self):
      return self.duenio

    def setEstado(self, nuevo_estado):
      self.estado = nuevo_estado


    def __str__(self):
        codigo_str = self.codigo_seguimiento if self.codigo_seguimiento is not None else "SIN-CÓDIGO"
        prioridad_str = "URGENTE" if self.es_urgente else "NORMAL"
        return (
            f"Paquete #{self.numero_llegada} | Guía: {codigo_str} | Prioridad: {prioridad_str} | "
            f"Estado: {self.estado} | Zona Destino: {self.codigo_zona} | "
            f"Cliente: {self.duenio.nombre_completo()}"
        )

class Cola:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def encolar(self, item):
        """Recibe un paquete al final de la fila de llegada."""
        self.items.append(item)

    def desencolar(self):
        """Atiende y saca al siguiente paquete en orden de llegada."""
        if not self.is_empty():
            return self.items.pop(0)
        else:
            return None

    def peek(self):
        """Permite ver quién sigue en la fila sin sacarlo."""
        if not self.is_empty():
            return self.items[0]
        else:
            return None

    def insertar_al_inicio(self, item):
        """Útil para el Mínimo 4 (Deshacer): devuelve un paquete al frente si se sacó por error."""
        self.items.insert(0, item)

    def size(self):
        return len(self.items)
```

### Mínimo 2 · Recepción y atención en orden de llegada
Necesitan: recibir un paquete, atender el siguiente y **ver quién sigue** sin sacarlo.

```python
# Tu código aquí: estructura para el orden de llegada

class ColaAtencion:
  def __init__(self):
    self.cola_urgente = Cola()
    self.cola_regulares = Cola()

  def encolar_paquete(self, paquete):
    if paquete.getUrgencia() == True:
      self.cola_urgente.encolar(paquete)
    else:
      self.cola_regulares.encolar(paquete)

  def despachar(self):
    if not self.cola_urgente.is_empty():
      return self.cola_urgente.desencolar()
    elif not self.cola_regulares.is_empty():
      return self.cola_regulares.desencolar()
    else:
      print("Amabas listas vacías")
      return None

  def ver_siguiente(self):
    if self.cola_urgente.peek() != None:
      return self.cola_urgente.peek()
    elif self.cola_regulares.peek() != None:
      return self.cola_regulares.peek()

  def esta_vacia(self):
    return self.cola_urgente.is_empty() and self.cola_regulares.is_empty()


```

### Mínimo 3 · Consulta por código de seguimiento (tabla hash propia)
Implementen la tabla hash completa: función hash, insertar, buscar y manejo de colisiones. **Sin `dict` ni `set`.**

```python
class TablaHash:
  def __init__(self):
    self.M = 101
    self.N = 0
    self.tabla = []

    for i in range(self.M):
      self.tabla.append([])

  def _hash(self, codigo):
    suma = 0
    for posicion, letra in enumerate(codigo):
      suma = suma + ord(letra)*31**posicion

    return suma % self.M

  def insertar(self, codigo_seguimiento, paquete):
    posicion = self._hash(codigo_seguimiento)
    self.tabla[posicion].append((codigo_seguimiento, paquete))
    self.N += 1

  def buscar(self, codigo_seguimiento):
    posicion = self._hash(codigo_seguimiento)
    comparaciones = 0
    lista = self.tabla[posicion]
    for i in lista:
      comparaciones += 1
      if i[0] == codigo_seguimiento:
        return (i[1], comparaciones)

    return (None, comparaciones)

  def factor_de_carga(self):
    return self.N/self.M
```

### Mínimo 4 · Deshacer acciones
Decidan qué acciones se registran y **qué información hay que guardar** para poder revertirlas.

```python
class PilaHistorial:
  def __init__(self):
    self.historial =[]

  def registrar_accion(self, paquete, estado_anterior, estado_nuevo):
    self.historial.append((paquete, estado_anterior, estado_nuevo))

  def deshacer(self):
    if len(self.historial) == 0:
      print("No hay acciones que deshacer")
      return False
    ultimo_movimiento = self.historial.pop()
    ultimo_movimiento[0].setEstado(ultimo_movimiento[1])
    print("El paquete ha vuelto a su estado anterior")
    return True
```

### Mínimo 5 · Búsqueda de zona por código (búsqueda binaria)
La lista de zonas debe estar **ordenada** antes de buscar.

```python
def busqueda_binaria(zonas, codigo_objetivo):
  inicio = 0
  final = len(zonas) - 1
  comparaciones = 0

  while inicio <= final:
    mitad = (inicio + final) // 2
    comparaciones += 1
    if zonas[mitad].getCodigoNumerico() == codigo_objetivo:
      return (zonas[mitad], comparaciones)
    elif zonas[mitad].getCodigoNumerico() < codigo_objetivo:
      inicio = mitad + 1
    else:
      final = mitad - 1

  return (None, comparaciones)


def insertar_zona_ordenada(zonas, nueva_zona):
  for i in range(len(zonas)):
    if nueva_zona.getCodigoNumerico() < zonas[i].getCodigoNumerico():
      zonas.insert(i, nueva_zona)
      return True

  zonas.append(nueva_zona)
```

### Mínimo 6 · Búsqueda de paquetes por nombre de cliente (búsqueda secuencial)
Piensen qué devuelve su búsqueda si hay varios paquetes con el mismo nombre.

```python
def buscar_paquete_cliente(paquetes, nombre_buscado):
  coincidencias = []
  comparaciones = 0
  for i in paquetes:
    comparaciones += 1
    if i.getDuenio().nombre_completo().lower().strip() == nombre_buscado.lower().strip(): coincidencias.append(i)

  return (coincidencias, comparaciones)
```

### Mínimo 8 · Medición de comparaciones
Midan cuántas comparaciones hace cada búsqueda (secuencial, binaria y hash) con **100, 1000 y 10000** datos y muestren una tabla con los resultados. Después escriban en la celda siguiente qué observan.

```python
def pruebas_rendimiento():
    print("\n--- modulo 5: pruebas de rendimiento ---")

    # la cabecera de tabla que pide el doc
    print("+---------+--------------------+-----------------+------------------+")
    print("|    N    | B. Secuencial O(N) | B. Binaria O(lg)| Tabla Hash O(1)  |")
    print("+---------+--------------------+-----------------+------------------+")

    tamanios = [100, 1000, 10000]

    for n in tamanios:
        paquetes = []
        zonas = []
        tabla = TablaHash()

        # agrandamos el limite de la tabla para la prueba de 10k
        # para que no colapse de colisiones y se note el O(1)
        tabla.M = n * 2
        tabla.tabla = []
        for _ in range(tabla.M):
            tabla.tabla.append([])
        tabla.N = 0

        # llenamos los datos falsos
        for i in range(1, n + 1):
            c = Cliente(f"Nombre{i}", f"Apellido{i}", str(i), "099999999", "Dir")
            codigo = f"GUIA-{i}"
            p = Paquete(i, c, i, "Destino", codigo)
            z = Zona(i, "Prov", "Cant", "Parr")

            paquetes.append(p)
            zonas.append(z)
            tabla.insertar(codigo, p)

        # buscamos siempre el ultimo (el N) para el peor caso
        cliente_buscar = f"Nombre{n} Apellido{n}"
        _, comp_sec = buscar_paquete_cliente(paquetes, cliente_buscar)

        _, comp_bin = busqueda_binaria(zonas, n)

        _, comp_hash = tabla.buscar(f"GUIA-{n}")

        # f-strings con numeros para cuadrar los espacios de la tabla
        print(f"| {n:7} | {comp_sec:18} | {comp_bin:15} | {comp_hash:16} |")

    print("+---------+--------------------+-----------------+------------------+")

# ejecutar
pruebas_rendimiento()
```

### Análisis de las mediciones
¿Qué pasa con las comparaciones al multiplicar los datos por 10? ¿Cuál búsqueda crece más rápido? ¿Su tabla hash se comportó como esperaban?

Descubrimos que al multiplicar por 10 la cantidad de datos, la búsqueda Secuencial es sin duda la que crece más rápido, al ser un crecimiento lineal esto tiene como consecuencia que si se multiplica la entrada por 10 así mismo se multiplicará el tiempo, haciendo que sea ineficiente para volúmenes grandes.

La búsqueda binaria demostró su eficiencia con su tiempo logarítmico, multiplicando la entrada por 10 apenas se sumaron unas 3 o 4 comparaciones cada vez ya que siempre la mitad será descartada.

Pero la tabla Hash resultó ser la mejor, ya que sin importar si buscábamos entre mil paquetes, el número de comparaciones siempre se mantuvo constante en 1, ya que gracias al hashing el algoritmo va directamente al casillero de memoria que le importa.
