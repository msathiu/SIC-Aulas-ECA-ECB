### Mínimo 1 · Modelado: paquete, cliente y zona
Definan las clases o estructuras. Antes de programar, escriban en papel los **sustantivos** y los **verbos** del problema.

```python
# Tu código aquí: clases Paquete, Cliente, Zona
'''
Contiene el nombre del cliente (de momento la unica información necesaria
para el sistema).
'''
class Cliente:
    def __init__(self, nombre: str):
        self.nombre = nombre

    def __str__(self):
        return f"Cliente: {self.nombre}"
```

```python
'''
Contiene el codigo de la zona y el nombre de la misma.
'''
class Zona:
    def __init__(self, codigo: int, nombre: str):
        self.codigo = codigo
        self.nombre = nombre

    def __str__(self):
        return f"Zona #{self.codigo} - {self.nombre}"
```

```python
'''
Contiene el un codigo que puede ser nulo, un cliente, una zona, un estado
de entregado (booleano) y una urgencia (booleano).
'''
class Paquete:
    def __init__(self, cliente: Cliente, zona: Zona, codigo: str = None, urgente: bool = False):
        self.codigo = codigo
        self.cliente = cliente
        self.zona = zona
        self.entregado = False
        self.urgente = urgente

    def __str__(self):
        estado_urgencia = "Urgente" if self.urgente else "Normal"
        estado_entrega = "Entregado" if self.entregado else "En bodega"
        return f"Paquete #{self.codigo} - Cliente: {self.cliente.nombre}, Zona: {self.zona.nombre}, Prioridad: {estado_urgencia}, Estado: {estado_entrega}"
```

```python
'''
Sera la encargada de almacenar los paquetes que seran entregados en orden de
llegada, asi como los paquetes urgentes.
'''
class ColaDePrioridad:
    def __init__(self):
        self.paquetes=[]  # en orden de llegada

    def _posicion_siguiente(self):
      # Busca el primer paquete urgente; si no hay ninguno, el que llegó primero
        for i, paquete in enumerate(self.paquetes):
            if paquete.urgente:
                return i
        return 0

    def push(self, paquete: Paquete):
        self.paquetes.append(paquete)

    def pop(self):
        if not self.paquetes:
            return None
        return self.paquetes.pop(self._posicion_siguiente())

    def peek(self):
        if not self.paquetes:
            return None
        return self.paquetes[self._posicion_siguiente()]

    def __str__(self):
        return str([str(p) for p in self.paquetes])
```

```python
'''
Implementación de la tabla Hash, su finalidad es almacenar los paquetes que
tengan codigo para poder encontrarlos rapidamente.
'''
class TablaHash:
    def __init__(self, capacidad=1009): # Número primo para reducir colisiones
        self.capacidad = capacidad
        self.tabla = [[] for _ in range(capacidad)] # Arreglo de listas (sin dict/set)
        self.comparaciones=0 # Contador para requisito minimo 8

    def _hash(self, clave) -> int:
        h=0
        for c in str(clave):
          h=(h*31+ord(c)) % self.capacidad
        return h

    def put(self, clave, paquete):
        clave=str(clave)
        bucket = self.tabla[self._hash(clave)]
        # Manejo de colisiones por encadenamiento
        for i, (k, p) in enumerate(bucket):
            if k == clave:
                bucket[i] = (clave, paquete) # Actualiza si ya existe
                return
        bucket.append((clave, paquete))

    def get(self, clave):
        self.comparaciones=0
        clave=str(clave)
        bucket = self.tabla[self._hash(clave)]
        for k, p in bucket:
            self.comparaciones+=1
            if k == clave:
                return p, self.comparaciones
        return None

    def __str__(self, limite=15):
        # Imprimimos las casillas ocupadas para mantener un orden visual
        ocupadas=[(i, b) for i, b in enumerate(self.tabla) if b]
        total=sum(len(b) for _, b in ocupadas)
        lineas=[f"TablaHash: {total} elementos en {len(ocupadas)} de"
                f"{self.capacidad} casillas"]
        for i, bucket in ocupadas[:limite]:
            claves=[k for k, _ in bucket]
            marca=" <-- colisión" if len(bucket) > 1 else ""
            lineas.append(f" casilla {i}: {claves}{marca}")
        if len(ocupadas)>limite:
            lineas.append(f" ... y más {len(ocupadas)-limite} casillas ocupadas más")
        return "\n".join(lineas)
```

```python
'''
Esta clase es el controlador del programa, los usuarios llamaran a
los metodos presestnes en esta para interactuar con el programa.
Ademas contiene dentro de si almacenada toda la información necesaria
para el correcto funcionamiento del sistema.
'''
class Paqueteria:

  def __init__(self):
        self.tabla_hash = TablaHash()
        self.cola_entregas = ColaDePrioridad()
        self.lista_sin_codigo = []
        self.lista_zonas = [] # Debe mantenerse ordenada
        self.historial_acciones = [] # Pila para deshacer

  def agregar_zona(self, zona: Zona):
        # Inserción ordenada para garantizar la Búsqueda Binaria
        if zona in self.lista_zonas:
          return
        self.lista_zonas.append(zona)
        self.lista_zonas.sort(key=lambda z: z.codigo)

  '''
  Agrega un paquete a la cola de prioridad (ademas en caso de tener
  codigo se agrega a la tabla hash, de lo contrario a la lista de paquetes
  por nombre).
  '''
  def agregar_paquete(self, paquete: Paquete):
        self.cola_entregas.push(paquete)
        if paquete.codigo:
            self.tabla_hash.put(paquete.codigo, paquete)
        else:
            self.lista_sin_codigo.append(paquete)
            # Clave sintética: nombre + posición[cite: 1]
            # clave_sintetica = f"{paquete.cliente.nombre}-{len(self.lista_sin_codigo)-1}"
            # self.tabla_hash.put(clave_sintetica, paquete)

  '''
  Devuelve el primer paquete de la cola de prioridad.
  '''
  def obtener_siguiente(self):
        return self.cola_entregas.peek()

  '''
  Entrega el primer paquete de la cola de prioridad y lo elimina de la misma.
  '''
  def entregar_siguiente(self):
        paquete = self.cola_entregas.pop()
        if paquete:
            self.historial_acciones.append(("ENTREGAR", paquete, paquete.entregado))
            paquete.entregado = True
        return paquete

  '''
  Cambia el estado de un paquete a entregado.
  '''
  def cambiar_entregado(self, paquete: Paquete):
        self.historial_acciones.append(("ESTADO", paquete, paquete.entregado))
        paquete.entregado = True

  '''
  Cambia la urgencia de un paquete a urgente.
  '''
  def cambiar_urgente(self, paquete: Paquete):
        # Según tus reglas, este cambio NO es revertible, no lo agregamos al historial
        paquete.urgente = True

  '''
  Cambia la zona de un paquete.
  '''
  def cambiar_zona(self, paquete: Paquete, codigo_zona: int):
        zona_nueva = self.encontrar_zona(codigo_zona)
        if zona_nueva:
            self.historial_acciones.append(("ZONA", paquete, paquete.zona))
            paquete.zona = zona_nueva

  '''
  Busca dentro de la tabla hash el paquete con el codigo proporcionado.
  '''
  def encontrar_codigo(self, codigo):
        # Búsqueda O(1) en Tabla Hash[cite: 1]
        return self.tabla_hash.get(codigo)

  '''
  Busca dentro de la lista de paquetes por nombre el cliente proporcionado.
  '''
  def encontrar_nombre(self, nombre):
        # Búsqueda Secuencial O(N)
        encontrados = []
        comparaciones = 0
        comparaciones_totales = 0
        for p in self.lista_sin_codigo:
            comparaciones += 1
            if p.cliente.nombre == nombre:
                comparaciones_totales = comparaciones
                encontrados.append(p)
        return encontrados, comparaciones_totales
  '''
  Busca dentro de la lista de zonas el codigo proporcionado.
  '''
  def encontrar_zona(self, codigo):
        # Búsqueda Binaria O(log N)
        inicio = 0
        fin = len(self.lista_zonas) - 1
        comparaciones = 0
        while inicio <= fin:
            comparaciones += 1
            medio = (inicio + fin) // 2
            if self.lista_zonas[medio].codigo == codigo:
                return self.lista_zonas[medio], comparaciones
            elif self.lista_zonas[medio].codigo < codigo:
                inicio = medio + 1
            else:
                fin = medio - 1
        return None

  '''
  Muestra el reporte de las acciones realizadas.
  '''
  def ver_reporte(self):
        print(f"Historial de acciones reversibles: {len(self.historial_acciones)}")

  '''
  Deshace la ultima acción realizada (en caso de que la misma no se pueda
  deshacer devuelve un mensaje).
  '''
  def deshacer_ultimo(self):
        if not self.historial_acciones:
            print("No hay acciones para deshacer.")
            return

        accion, paquete, valor_previo = self.historial_acciones.pop()
        if accion == "ESTADO" or accion == "ENTREGAR":
            paquete.entregado = valor_previo
            print(f"Deshecho: Estado de paquete vuelto a {valor_previo}")
        elif accion == "ZONA":
            paquete.zona = valor_previo
            print(f"Deshecho: Zona de paquete devuelta a {valor_previo.nombre}")

  '''
  Devuelve la tabla hash con los paquetes.
  '''
  def get_tabla_paquetes(self):
        return self.tabla_hash

  '''
  Devuelve la lista de zonas.
  '''
  def get_zonas(self):
        return self.lista_zonas
```

```python
# Tu código aquí: mediciones
import math
import time
from random import randint

print("=== MEDICIÓN DE COMPARACIONES Y TIEMPOS ===")
print(f"{'N Datos':<10} | {'Buscado':<10} | {'Secuencial':<22} | {'Binaria':<18} | {'Hash':<18}")
print("-" * 88)

tamaños = [100, 1000, 10000]

for n in tamaños:
    paqueteria_prueba = Paqueteria()
    for i in range(n):
        cliente_prueba = Cliente("Cliente " + str(i))
        zona_prueba = Zona(i, "Zona " + str(i))

        paquete_prueba_codigo = Paquete(cliente_prueba, zona_prueba, str(i), (i % 2 == 0))
        paquete_prueba_nombre = Paquete(cliente_prueba, zona_prueba, urgente=(i % 2 == 0))

        paqueteria_prueba.agregar_paquete(paquete_prueba_codigo)
        paqueteria_prueba.agregar_paquete(paquete_prueba_nombre)
        paqueteria_prueba.agregar_zona(zona_prueba)

    numero_encontrar = randint(1, n)

    # --- 1. Búsqueda Secuencial (Por nombre) ---
    inicio = time.perf_counter()
    paquetes, comp_secuencial = paqueteria_prueba.encontrar_nombre(f"Cliente {numero_encontrar}")
    tiempo_secuencial = time.perf_counter() - inicio

    # --- 2. Búsqueda Binaria (Zona) ---
    inicio = time.perf_counter()
    paquete, comp_binaria = paqueteria_prueba.encontrar_zona(numero_encontrar)
    tiempo_binaria = time.perf_counter() - inicio

    # --- 3. Búsqueda Hash (Por código) ---
    inicio = time.perf_counter()
    paquete, comp_hash = paqueteria_prueba.encontrar_codigo(str(numero_encontrar))
    tiempo_hash = time.perf_counter() - inicio

    # Imprimir fila de comparaciones incluyendo el número buscado
    print(f"{n:<10} | {numero_encontrar:<10} | {str(comp_secuencial) + ' comp':<22} | {str(comp_binaria) + ' comp':<18} | {str(comp_hash) + ' comp':<18}")
    # Imprimir fila de tiempos
    print(f"{'Tiempo (s)':<10} | {'':<10} | {tiempo_secuencial:<22.6f} | {tiempo_binaria:<18.6f} | {tiempo_hash:<18.6f}")
    print("-" * 88)
```
